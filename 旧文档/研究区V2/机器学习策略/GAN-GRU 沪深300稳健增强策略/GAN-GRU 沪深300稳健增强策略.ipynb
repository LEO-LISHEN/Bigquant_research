# GAN-GRU 沪深300稳健增强策略

**GAN-GRU 沪深300稳健增强策略**是一套基于机器学习的指数增强策略，主要以沪深300成分股为投资范围，先根据个股历史行情数据构造动量、均线偏离、波动率、成交量、RSI、MACD、WR 等量价技术因子，再通过 GAN 模块对因子序列进行特征增强和类降噪处理，使后续的 StockGRU 模型能够更好地学习股票过去一段时间的量价变化与未来收益之间的关系。策略每周五进行一次调仓，根据模型预测分数将沪深300成分股分为五组，在原始指数权重基础上对高分股票小幅超配、对低分股票小幅低配，从而在控制组合整体接近沪深300的同时，尝试获取相对于指数的超额收益。


## 原版模型策略

```python
# -*- coding: utf-8 -*-
"""
GAN + GRU 沪深300指数增强策略
BigQuant DAI + BigTrader 原生日频回测版

复刻原则：
1. 不修改原策略的神经网络结构与超参数；
2. 不修改20个输入特征及其计算方式；
3. 不修改训练期、40日窗口和 target = pct_chg.shift(-5)；
4. 不修改样本标签索引 y = targets[idx + SEQ_LEN]；
5. 不修改“Generator生成训练输入、实时预测直接输入标准化原特征”的原逻辑；
6. 保留每周五调仓及五组绝对权重偏移规则；不再要求14:40精确成交；
7. 不增加ST、停牌、涨跌停、换手率或单票权重等额外筛选约束；
8. 只将聚宽数据/交易接口替换为BigQuant官方DAI与BigTrader原生接口。

BigQuant官方接口依据：
- BigTrader原生回测：from bigquant import bigtrader, dai；bigtrader.run(...)
- 日频回测：frequency=bigtrader.Frequency.DAILY
- 每个日线Bar触发handle_data；周五执行调仓
- 目标权重下单：context.order_target_percent(instrument, target_weight)
- 股票日线：cn_stock_bar1d；成交额字段为amount
- cn_stock_bar1d为后复权行情。为贴近原代码fq='pre'，本代码使用adjust_factor
  将每次训练/预测所用OHLC转换为“以该次数据末日为基准”的前复权口径。

注意：
- 原策略自身存在的目标定义、训练/推理输入不一致等问题，本版刻意保留；
- BigQuant平台需具备PyTorch、cn_stock_bar1d、cn_stock_prefactors以及
  cn_stock_index_weight相应数据权限；
- 如果官方指数权重表不可读取，按照原版google_get_index_weights的逻辑，
  自动退化为当期沪深300成分股等权。
"""

from datetime import datetime
import gc
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# 当前BigQuant DAI环境不提供dai.get_datasource_schema，
# 因此本策略只使用官方文档明确给出的固定表名和字段，不做运行时schema探测。


# =============================================================================
# 1. 参数：与原版保持一致
# =============================================================================

INDEX_CODE = "000300.SH"

TRAIN_START = "2023-01-01"
TRAIN_END = "2024-05-01"

# 原代码未包含回测区间；BigTrader必须显式给出，因此从训练期结束后的下一交易周开始。
# 可只修改这三个回测参数，不影响策略逻辑。
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2024-08-29"
CAPITAL_BASE = 1_000_000

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5

GAN_EPOCHS = 4
GRU_EPOCHS = 8
BATCH_SIZE = 1024
BIAS = 0.008

DAILY_LOOKBACK_COUNT = SEQ_LEN + 60  # 原版实时端count=100

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

FEATURE_COLUMNS = [
    "open", "close", "high", "low", "volume", "money",
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5",
    "std_5", "std_20", "rsi_14", "macd_hist", "wr_14",
    "volume_ratio",
]


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)


# =============================================================================
# 4. 20个特征：逐项复刻原版
# =============================================================================

def calculate_all_factors(df):
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)

    # 动量与收益率特征
    df["pct_chg"] = df.groupby("instrument")["close"].pct_change()
    df["mom_5"] = df.groupby("instrument")["close"].pct_change(5)
    df["mom_10"] = df.groupby("instrument")["close"].pct_change(10)
    df["mom_20"] = df.groupby("instrument")["close"].pct_change(20)

    # 均线与距离特征
    df["ma5"] = df.groupby("instrument")["close"].transform(
        lambda x: x.rolling(5).mean()
    )
    df["ma10"] = df.groupby("instrument")["close"].transform(
        lambda x: x.rolling(10).mean()
    )
    df["ma20"] = df.groupby("instrument")["close"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    # 波动率与量价特征
    df["vol_ma5"] = df.groupby("instrument")["volume"].transform(
        lambda x: x.rolling(5).mean()
    )
    df["std_5"] = df.groupby("instrument")["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument")["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    # RSI 14
    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = df.groupby("instrument")["close"].transform(
        lambda x: get_rsi(x, 14)
    )

    # MACD柱状图
    df["ema12"] = df.groupby("instrument")["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = df.groupby("instrument")["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument")["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist"] = (df["dif"] - df["dea"]) * 2

    # WR 14
    df["high_14"] = df.groupby("instrument")["high"].transform(
        lambda x: x.rolling(14).max()
    )
    df["low_14"] = df.groupby("instrument")["low"].transform(
        lambda x: x.rolling(14).min()
    )
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    feat_cols = FEATURE_COLUMNS.copy()

    df.drop(
        columns=["ema12", "ema26", "dif", "dea", "high_14", "low_14"],
        errors="ignore",
        inplace=True,
    )
    return df, feat_cols


# =============================================================================
# 5. 模型训练：逐项复刻原版
# =============================================================================

def train_and_save_models(context, train_stocks):
    start_fetch = (
        pd.to_datetime(TRAIN_START) - pd.Timedelta(days=40)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"开始训练：股票数={len(train_stocks)}，"
        f"行情区间={start_fetch}~{TRAIN_END}"
    )

    price_df = _query_daily_bars(
        train_stocks,
        start_fetch,
        TRAIN_END,
    )
    if price_df.empty:
        raise RuntimeError("训练行情为空")

    # 对应原版 get_price(..., fq='pre')。
    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df, feat_cols = calculate_all_factors(price_df)

    context.feat_dim = len(feat_cols)
    price_df["target"] = price_df.groupby("instrument")["pct_chg"].shift(-5)

    price_df = price_df[
        (price_df["date"] >= pd.Timestamp(TRAIN_START))
        & (price_df["date"] <= pd.Timestamp(TRAIN_END))
    ]
    price_df.dropna(inplace=True)

    x_list = []
    y_list = []

    for _, group in price_df.groupby("instrument"):
        if len(group) < SEQ_LEN:
            continue
        feats = group[feat_cols].values
        targets = group["target"].values
        for idx in range(len(group) - SEQ_LEN):
            x_list.append(feats[idx: idx + SEQ_LEN])
            y_list.append(targets[idx + SEQ_LEN])

    if not x_list:
        raise RuntimeError("未构造出有效训练样本")

    x_tensor = torch.tensor(np.array(x_list, dtype=np.float32))
    y_tensor = torch.tensor(
        np.array(y_list, dtype=np.float32).reshape(-1, 1)
    )

    del x_list, y_list
    gc.collect()

    x_flat = x_tensor.view(-1, context.feat_dim)
    context.train_mean = x_flat.mean(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1)
    context.train_std = x_flat.std(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1) + 1e-8
    x_tensor = (x_tensor - context.train_mean) / context.train_std

    train_loader = DataLoader(
        TensorDataset(x_tensor, y_tensor),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )

    # GAN
    generator = Generator(context.feat_dim, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        context.feat_dim,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()

        for real_data, _ in train_loader:
            real_data = real_data.to(DEVICE)
            bs = real_data.size(0)
            real_label = torch.ones(bs, device=DEVICE)
            fake_label = torch.zeros(bs, device=DEVICE)

            discriminator.zero_grad()
            loss_d = (
                criterion(discriminator(real_data), real_label)
                + criterion(
                    discriminator(generator(real_data).detach()),
                    fake_label,
                )
            )
            loss_d.backward()
            opt_d.step()

            generator.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion(discriminator(fake_data), real_label)
                + LAMBDA_REC * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

        context.logger.info(f"GAN训练完成：epoch={epoch + 1}/{GAN_EPOCHS}")

    # 分批特征增强：与原版一样，仅使用Generator输出训练GRU，不拼接原样本。
    generator.eval()
    with torch.no_grad():
        x_enhanced = torch.cat(
            [
                generator(x_tensor[i: i + 1024].to(DEVICE)).cpu()
                for i in range(0, x_tensor.size(0), 1024)
            ],
            dim=0,
        )

    del x_tensor, x_flat, price_df
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # GRU
    enhanced_loader = DataLoader(
        TensorDataset(x_enhanced, y_tensor),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )

    context.gru_model = StockGRU(context.feat_dim, 64, 2).to(DEVICE)
    optimizer_gru = optim.Adam(context.gru_model.parameters(), lr=0.001)
    criterion_gru = nn.MSELoss()

    for epoch in range(GRU_EPOCHS):
        context.gru_model.train()
        for batch_x, batch_y in enhanced_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)
            optimizer_gru.zero_grad()
            criterion_gru(context.gru_model(batch_x), batch_y).backward()
            optimizer_gru.step()

        context.logger.info(f"GRU训练完成：epoch={epoch + 1}/{GRU_EPOCHS}")

    context.gru_model.eval()
    context.model_ready = True

    del x_enhanced, y_tensor, generator, discriminator
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.logger.info("GAN + GRU模型已就绪")


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None

    # 权重数据在bigtrader.run之前只读取一次，同时用于限定回测股票池。
    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.logger.info(
        f"已加载沪深300权重缓存：{len(context.index_weights_df):,}行，"
        f"股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    # 训练股票池：训练结束日的沪深300成分股。
    train_weights = _weights_asof(context.index_weights_df, TRAIN_END)
    train_stocks = train_weights["instrument"].tolist()
    if not train_stocks:
        raise RuntimeError(f"无法取得{TRAIN_END}沪深300成分股")

    train_and_save_models(context, train_stocks)

    # 一次性读取回测期可能用到的全部日线原始数据；每周预测时仍按原版逻辑
    # 截取当期股票最近100个完整日线，并重新做末端前复权与因子计算。
    backtest_sections = context.index_weights_df[
        (context.index_weights_df["date"] >= pd.Timestamp(context.start_date) - pd.Timedelta(days=180))
        & (context.index_weights_df["date"] <= pd.Timestamp(context.end_date))
    ]
    universe = sorted(
        backtest_sections["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300股票并集为空")

    daily_start = (
        pd.Timestamp(context.start_date) - pd.Timedelta(days=220)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"读取回测期日线缓存：{daily_start}~{context.end_date}，"
        f"股票并集={len(universe)}"
    )
    daily_df = _query_daily_bars(
        universe,
        daily_start,
        context.end_date,
    )
    if daily_df.empty:
        raise RuntimeError("回测期日线缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }

    del daily_df, backtest_sections
    gc.collect()


def _build_realtime_factor_data(context, stocks, current_date):
    """
    复刻原版：
        get_price(stocks, end_date=current_date,
                  count=SEQ_LEN+60, frequency='daily', fq='pre')

    日频回测在周五调仓，只使用current_date之前已经完整结束的日线。
    因此信号在周五交易前已经可得，不使用周五收盘后信息。
    """
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue

        # 原版count=100；从完整日线中取最近100条。
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if section.empty:
            continue
        pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df, feat_cols = calculate_all_factors(realtime_df)
    realtime_df.dropna(inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(
        context,
        stocks,
        current_date,
    )
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    # 批量推理只减少Python循环和模型调用次数，不改变每只股票的输入、
    # 标准化方式、模型结构或预测结果含义。实时推理仍不经过Generator。
    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )

    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 回测静态数据准备
# =============================================================================

def _prepare_backtest_static_data():
    """
    在启动BigTrader前读取一次指数权重，并据此限定回测股票池。

    这样日频引擎只加载回测期沪深300历史成分股并集，避免在未指定
    instruments时准备全市场行情。
    """
    weight_start = (
        min(pd.Timestamp(TRAIN_END), pd.Timestamp(BACKTEST_START))
        - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        using_official = False

    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    return weights, using_official, universe


(
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
) = _prepare_backtest_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

performance.render()
```

## 改进版模型1

1. 修改预测标签

原模型预测的是未来第5个交易日的单日收益率，现改为未来5个交易日的累计收益率，使预测目标与每周调仓周期更加一致。

2. 修正样本窗口索引

将40日特征窗口的最后一天与未来5日累计收益标签直接对齐，消除了原代码中额外错开一个交易日的问题。

3. 调整GAN样本使用方式

原模型只使用GAN生成样本训练GRU，现改为使用原始标准化样本和GAN生成样本共同训练。这样既保留原始数据的信息，也能够利用GAN进行数据增强。

4. 优化量价特征

将绝对价格、成交量和成交额改为相对特征，例如开盘跳空、日内收益、相对最高价、相对最低价、相对成交量和相对成交额，以提高不同股票之间的可比性，并降低绝对数值规模对模型的影响。

除上述改动外，GAN和GRU结构、40日输入窗口、训练区间、每周调仓以及沪深300权重调整逻辑均保持不变。

```python
# -*- coding: utf-8 -*-
"""
GAN + GRU 沪深300指数增强策略
BigQuant DAI + BigTrader 原生日频回测改进版

本版在原日频移植版本基础上完成三项核心修改：
1. 标签改为窗口末日到未来第5个交易日的累计收益率：
       future_return_5 = close.shift(-5) / close - 1
2. 修正样本索引：每个40日窗口直接匹配窗口最后一日对应的future_return_5；
3. GRU使用“原始标准化样本 + Generator生成样本”共同训练，不再用生成样本替代原始样本；
4. 将直接进入模型的绝对价格、成交量、成交额改为相对特征，并同步消除
   5日均量和MACD柱的绝对尺度。

保持不变的部分：
- 固定训练期、GAN/GRU网络结构和主要超参数；
- 40日输入窗口；
- 每周五调仓；
- 沪深300官方权重基础上的五组绝对权重偏移规则；
- BigQuant DAI数据接口与BigTrader原生日频回测引擎；
- 实时预测直接输入GRU，不经过Generator。

BigQuant接口：
- from bigquant import bigtrader, dai
- bigtrader.run(..., frequency=bigtrader.Frequency.DAILY)
- cn_stock_bar1d日线数据，amount映射为成交额money
- cn_stock_index_weight指数权重
- context.order_target_percent目标仓位下单
"""

from datetime import datetime
import gc
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import ConcatDataset, DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# 当前BigQuant DAI环境不提供dai.get_datasource_schema，
# 因此本策略只使用官方文档明确给出的固定表名和字段，不做运行时schema探测。


# =============================================================================
# 1. 参数：与原版保持一致
# =============================================================================

INDEX_CODE = "000300.SH"

TRAIN_START = "2023-01-01"
TRAIN_END = "2024-05-01"

# 原代码未包含回测区间；BigTrader必须显式给出，因此从训练期结束后的下一交易周开始。
# 可只修改这三个回测参数，不影响策略逻辑。
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2024-08-29"
CAPITAL_BASE = 1_000_000

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5

GAN_EPOCHS = 4
GRU_EPOCHS = 8
BATCH_SIZE = 1024
BIAS = 0.008

DAILY_LOOKBACK_COUNT = SEQ_LEN + 60  # 原版实时端count=100

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

FEATURE_COLUMNS = [
    # 六个平稳化基础量价特征
    "open_gap", "intraday_ret", "high_rel", "low_rel",
    "volume_rel20", "money_rel20",
    # 收益、动量、均线、波动和技术指标
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5_rel20",
    "std_5", "std_20", "rsi_14", "macd_hist_rel", "wr_14",
    "volume_ratio",
]


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)


# =============================================================================
# 4. 20个特征：相对化和平稳化版本
# =============================================================================

def calculate_all_factors(df):
    """
    构造20维量价时序特征。

    关键调整：
    - 不再把绝对open/close/high/low直接输入模型；
    - 不再把绝对volume/money或绝对5日均量直接输入模型；
    - MACD柱除以收盘价，消除股票价格数量级差异。
    """
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)
    grouped = df.groupby("instrument", sort=False)

    # -------------------------------------------------------------------------
    # 六个平稳化基础量价特征
    # -------------------------------------------------------------------------
    df["prev_close"] = grouped["close"].shift(1)

    # 相对前收盘的开盘跳空、最高价和最低价；日内收盘收益反映K线实体。
    df["open_gap"] = df["open"] / (df["prev_close"] + 1e-8) - 1.0
    df["intraday_ret"] = df["close"] / (df["open"] + 1e-8) - 1.0
    df["high_rel"] = df["high"] / (df["prev_close"] + 1e-8) - 1.0
    df["low_rel"] = df["low"] / (df["prev_close"] + 1e-8) - 1.0

    # 成交量和成交额相对各自20日均值，消除股票规模及年代数量级差异。
    df["volume_ma20"] = grouped["volume"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["money_ma20"] = grouped["money"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["volume_rel20"] = df["volume"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["money_rel20"] = df["money"] / (df["money_ma20"] + 1e-8) - 1.0

    # -------------------------------------------------------------------------
    # 收益率与动量特征
    # -------------------------------------------------------------------------
    df["pct_chg"] = grouped["close"].pct_change()
    df["mom_5"] = grouped["close"].pct_change(5)
    df["mom_10"] = grouped["close"].pct_change(10)
    df["mom_20"] = grouped["close"].pct_change(20)

    # -------------------------------------------------------------------------
    # 均线距离特征
    # -------------------------------------------------------------------------
    df["ma5"] = grouped["close"].transform(lambda x: x.rolling(5).mean())
    df["ma10"] = grouped["close"].transform(lambda x: x.rolling(10).mean())
    df["ma20"] = grouped["close"].transform(lambda x: x.rolling(20).mean())
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    # -------------------------------------------------------------------------
    # 波动率与量价趋势特征
    # -------------------------------------------------------------------------
    df["vol_ma5"] = grouped["volume"].transform(lambda x: x.rolling(5).mean())
    df["vol_ma5_rel20"] = df["vol_ma5"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["std_5"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    # RSI 14
    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = grouped["close"].transform(lambda x: get_rsi(x, 14))

    # MACD柱状图：除以收盘价，转换成跨股票可比的相对尺度。
    df["ema12"] = grouped["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = grouped["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument", sort=False)["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist_rel"] = ((df["dif"] - df["dea"]) * 2) / (
        df["close"].abs() + 1e-8
    )

    # WR 14
    df["high_14"] = grouped["high"].transform(lambda x: x.rolling(14).max())
    df["low_14"] = grouped["low"].transform(lambda x: x.rolling(14).min())
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    # 对除零或异常行情产生的无穷值统一转为缺失值，由后续dropna剔除。
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    feat_cols = FEATURE_COLUMNS.copy()
    df.drop(
        columns=[
            "prev_close", "volume_ma20", "money_ma20",
            "ma5", "ma10", "ma20", "vol_ma5",
            "ema12", "ema26", "dif", "dea", "high_14", "low_14",
        ],
        errors="ignore",
        inplace=True,
    )
    return df, feat_cols


# =============================================================================
# 5. 模型训练：未来5日累计收益 + 原始/生成样本联合训练
# =============================================================================

def train_and_save_models(context, train_stocks):
    start_fetch = (
        pd.to_datetime(TRAIN_START) - pd.Timedelta(days=40)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"开始训练：股票数={len(train_stocks)}，"
        f"行情区间={start_fetch}~{TRAIN_END}"
    )

    price_df = _query_daily_bars(
        train_stocks,
        start_fetch,
        TRAIN_END,
    )
    if price_df.empty:
        raise RuntimeError("训练行情为空")

    # 对应原版 get_price(..., fq='pre')。
    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df, feat_cols = calculate_all_factors(price_df)

    context.feat_dim = len(feat_cols)

    # 标签：以当前行收盘价为起点，到未来第5个交易日收盘价的累计收益率。
    # target[t] = close[t+5] / close[t] - 1
    grouped_close = price_df.groupby("instrument", sort=False)["close"]
    price_df["target"] = grouped_close.shift(-5) / price_df["close"] - 1.0

    price_df = price_df[
        (price_df["date"] >= pd.Timestamp(TRAIN_START))
        & (price_df["date"] <= pd.Timestamp(TRAIN_END))
    ]
    price_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    price_df.dropna(subset=feat_cols + ["target"], inplace=True)

    x_list = []
    y_list = []

    for _, group in price_df.groupby("instrument", sort=False):
        group = group.sort_values("date").reset_index(drop=True)
        if len(group) < SEQ_LEN:
            continue

        feats = group[feat_cols].to_numpy(dtype=np.float32)
        targets = group["target"].to_numpy(dtype=np.float32)

        # 窗口最后一日end_idx直接匹配同一日的未来5日累计收益标签。
        # 例如输入[0, ..., 39]，标签就是target[39]，不存在额外一日错位。
        for end_idx in range(SEQ_LEN - 1, len(group)):
            start_idx = end_idx - SEQ_LEN + 1
            x_list.append(feats[start_idx: end_idx + 1])
            y_list.append(targets[end_idx])

    if not x_list:
        raise RuntimeError("未构造出有效训练样本")

    x_tensor = torch.tensor(np.array(x_list, dtype=np.float32))
    y_tensor = torch.tensor(
        np.array(y_list, dtype=np.float32).reshape(-1, 1)
    )

    del x_list, y_list
    gc.collect()

    x_flat = x_tensor.view(-1, context.feat_dim)
    context.train_mean = x_flat.mean(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1)
    context.train_std = x_flat.std(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1) + 1e-8
    x_tensor = (x_tensor - context.train_mean) / context.train_std

    train_loader = DataLoader(
        TensorDataset(x_tensor, y_tensor),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )

    # GAN
    generator = Generator(context.feat_dim, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        context.feat_dim,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()

        for real_data, _ in train_loader:
            real_data = real_data.to(DEVICE)
            bs = real_data.size(0)
            real_label = torch.ones(bs, device=DEVICE)
            fake_label = torch.zeros(bs, device=DEVICE)

            discriminator.zero_grad()
            loss_d = (
                criterion(discriminator(real_data), real_label)
                + criterion(
                    discriminator(generator(real_data).detach()),
                    fake_label,
                )
            )
            loss_d.backward()
            opt_d.step()

            generator.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion(discriminator(fake_data), real_label)
                + LAMBDA_REC * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

        context.logger.info(f"GAN训练完成：epoch={epoch + 1}/{GAN_EPOCHS}")

    # 分批生成增强样本。原始标准化样本与生成样本共同训练GRU。
    generator.eval()
    with torch.no_grad():
        x_enhanced = torch.cat(
            [
                generator(x_tensor[i: i + 1024].to(DEVICE)).cpu()
                for i in range(0, x_tensor.size(0), 1024)
            ],
            dim=0,
        )

    # 使用ConcatDataset避免再复制一份大型X张量；两个数据集共享同一组标签。
    original_dataset = TensorDataset(x_tensor, y_tensor)
    enhanced_dataset = TensorDataset(x_enhanced, y_tensor)
    joint_dataset = ConcatDataset([original_dataset, enhanced_dataset])
    joint_loader = DataLoader(
        joint_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
    )

    del x_flat, price_df
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # GRU：每轮同时遍历原始样本和Generator生成样本。


    context.gru_model = StockGRU(context.feat_dim, 64, 2).to(DEVICE)
    optimizer_gru = optim.Adam(context.gru_model.parameters(), lr=0.001)
    criterion_gru = nn.MSELoss()

    for epoch in range(GRU_EPOCHS):
        context.gru_model.train()
        for batch_x, batch_y in joint_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)
            optimizer_gru.zero_grad()
            criterion_gru(context.gru_model(batch_x), batch_y).backward()
            optimizer_gru.step()

        context.logger.info(f"GRU训练完成：epoch={epoch + 1}/{GRU_EPOCHS}")

    context.gru_model.eval()
    context.model_ready = True

    del x_tensor, x_enhanced, y_tensor, original_dataset, enhanced_dataset, joint_dataset, generator, discriminator
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.logger.info("GAN + GRU模型已就绪：目标=未来5日累计收益，GRU训练集=原始样本+生成样本")


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None

    # 权重数据在bigtrader.run之前只读取一次，同时用于限定回测股票池。
    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.logger.info(
        f"已加载沪深300权重缓存：{len(context.index_weights_df):,}行，"
        f"股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    # 训练股票池：训练结束日的沪深300成分股。
    train_weights = _weights_asof(context.index_weights_df, TRAIN_END)
    train_stocks = train_weights["instrument"].tolist()
    if not train_stocks:
        raise RuntimeError(f"无法取得{TRAIN_END}沪深300成分股")

    train_and_save_models(context, train_stocks)

    # 一次性读取回测期可能用到的全部日线原始数据；每周预测时仍按原版逻辑
    # 截取当期股票最近100个完整日线，并重新做末端前复权与因子计算。
    backtest_sections = context.index_weights_df[
        (context.index_weights_df["date"] >= pd.Timestamp(context.start_date) - pd.Timedelta(days=180))
        & (context.index_weights_df["date"] <= pd.Timestamp(context.end_date))
    ]
    universe = sorted(
        backtest_sections["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300股票并集为空")

    daily_start = (
        pd.Timestamp(context.start_date) - pd.Timedelta(days=220)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"读取回测期日线缓存：{daily_start}~{context.end_date}，"
        f"股票并集={len(universe)}"
    )
    daily_df = _query_daily_bars(
        universe,
        daily_start,
        context.end_date,
    )
    if daily_df.empty:
        raise RuntimeError("回测期日线缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }

    del daily_df, backtest_sections
    gc.collect()


def _build_realtime_factor_data(context, stocks, current_date):
    """
    复刻原版：
        get_price(stocks, end_date=current_date,
                  count=SEQ_LEN+60, frequency='daily', fq='pre')

    日频回测在周五调仓，只使用current_date之前已经完整结束的日线。
    因此信号在周五交易前已经可得，不使用周五收盘后信息。
    """
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue

        # 原版count=100；从完整日线中取最近100条。
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if section.empty:
            continue
        pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df, feat_cols = calculate_all_factors(realtime_df)
    realtime_df.dropna(inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(
        context,
        stocks,
        current_date,
    )
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    # 批量推理只减少Python循环和模型调用次数，不改变每只股票的输入、
    # 标准化方式、模型结构或预测结果含义。预测值对应未来5日累计收益，实时推理仍不经过Generator。
    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )

    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 回测静态数据准备
# =============================================================================

def _prepare_backtest_static_data():
    """
    在启动BigTrader前读取一次指数权重，并据此限定回测股票池。

    这样日频引擎只加载回测期沪深300历史成分股并集，避免在未指定
    instruments时准备全市场行情。
    """
    weight_start = (
        min(pd.Timestamp(TRAIN_END), pd.Timestamp(BACKTEST_START))
        - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        using_official = False

    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    return weights, using_official, universe


(
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
) = _prepare_backtest_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

```

## 改进版模型2

本版本在改进模型1基础上增加了严格的时间验证集和基于 RankIC 的早停机制。

1. 时间顺序划分数据

训练样本不再全部用于模型训练，而是按照日期顺序划分为训练集和验证集。较早的数据用于训练，较晚的数据用于验证，避免随机划分造成相邻时序窗口泄漏。

2. 设置隔离期

训练集和验证集之间预留5个交易日，防止训练样本的未来5日收益标签使用到验证期行情。

3. 防止验证集信息泄漏

特征均值和标准差仅使用训练集计算，GAN也只使用训练集训练。验证集不会参与模型参数更新，仅用于样本外评估。

4. 使用 RankIC 评价模型

每轮GRU训练结束后，模型会在验证集上预测，并按日期计算预测分数与实际未来5日收益之间的横截面 RankIC。

RankIC越高，表示模型对股票未来收益排序的预测能力越强，与策略实际按照预测分数选股和分组的逻辑更加一致。

5. RankIC早停

GRU不再固定训练8轮，而是设置最大训练轮数。若验证集RankIC连续多轮没有改善，则提前停止训练。

最终使用的是验证集平均RankIC最高的那一轮模型，而不是最后一轮模型，从而降低过拟合风险。

除上述改动外，GAN与GRU结构、未来5日累计收益标签、原始样本与生成样本联合训练、相对量价特征以及组合调仓逻辑均保持不变。

```python
# -*- coding: utf-8 -*-
"""
GAN + GRU 沪深300指数增强策略
BigQuant DAI + BigTrader 原生日频回测改进版

本版在原日频移植版本基础上完成三项核心修改：
1. 标签改为窗口末日到未来第5个交易日的累计收益率：
       future_return_5 = close.shift(-5) / close - 1
2. 修正样本索引：每个40日窗口直接匹配窗口最后一日对应的future_return_5；
3. GRU使用“原始标准化样本 + Generator生成样本”共同训练，不再用生成样本替代原始样本；
4. 将直接进入模型的绝对价格、成交量、成交额改为相对特征，并同步消除
   5日均量和MACD柱的绝对尺度；
5. 按样本日期严格划分训练集与验证集，在两者之间留出5个交易日隔离期；
6. GAN仅使用训练集，GRU每轮后计算验证集逐日横截面RankIC，并按RankIC早停。

保持不变的部分：
- 固定训练期、GAN/GRU网络结构和主要超参数；
- 40日输入窗口；
- 每周五调仓；
- 沪深300官方权重基础上的五组绝对权重偏移规则；
- BigQuant DAI数据接口与BigTrader原生日频回测引擎；
- 实时预测直接输入GRU，不经过Generator。

BigQuant接口：
- from bigquant import bigtrader, dai
- bigtrader.run(..., frequency=bigtrader.Frequency.DAILY)
- cn_stock_bar1d日线数据，amount映射为成交额money
- cn_stock_index_weight指数权重
- context.order_target_percent目标仓位下单
"""

from datetime import datetime
import gc
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import ConcatDataset, DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# 当前BigQuant DAI环境不提供dai.get_datasource_schema，
# 因此本策略只使用官方文档明确给出的固定表名和字段，不做运行时schema探测。


# =============================================================================
# 1. 参数：与原版保持一致
# =============================================================================

INDEX_CODE = "000300.SH"

TRAIN_START = "2023-01-01"
TRAIN_END = "2024-05-01"

# 原代码未包含回测区间；BigTrader必须显式给出，因此从训练期结束后的下一交易周开始。
# 可只修改这三个回测参数，不影响策略逻辑。
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2024-08-29"
CAPITAL_BASE = 1_000_000

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5

GAN_EPOCHS = 4

# GRU采用验证集RankIC早停，不再固定使用最后一轮。
GRU_MAX_EPOCHS = 20
EARLY_STOP_PATIENCE = 4
RANKIC_MIN_DELTA = 1e-4

# 严格时间验证：最后20%的样本日期作为验证期，
# 验证期前留出5个交易日，防止未来5日收益标签跨越训练/验证边界。
VALIDATION_RATIO = 0.20
PURGE_TRADING_DAYS = 5
MIN_VALIDATION_DATES = 20
MIN_RANKIC_STOCKS = 30

BATCH_SIZE = 1024
BIAS = 0.008

DAILY_LOOKBACK_COUNT = SEQ_LEN + 60  # 原版实时端count=100

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

FEATURE_COLUMNS = [
    # 六个平稳化基础量价特征
    "open_gap", "intraday_ret", "high_rel", "low_rel",
    "volume_rel20", "money_rel20",
    # 收益、动量、均线、波动和技术指标
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5_rel20",
    "std_5", "std_20", "rsi_14", "macd_hist_rel", "wr_14",
    "volume_ratio",
]


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)


# =============================================================================
# 4. 20个特征：相对化和平稳化版本
# =============================================================================

def calculate_all_factors(df):
    """
    构造20维量价时序特征。

    关键调整：
    - 不再把绝对open/close/high/low直接输入模型；
    - 不再把绝对volume/money或绝对5日均量直接输入模型；
    - MACD柱除以收盘价，消除股票价格数量级差异。
    """
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)
    grouped = df.groupby("instrument", sort=False)

    # -------------------------------------------------------------------------
    # 六个平稳化基础量价特征
    # -------------------------------------------------------------------------
    df["prev_close"] = grouped["close"].shift(1)

    # 相对前收盘的开盘跳空、最高价和最低价；日内收盘收益反映K线实体。
    df["open_gap"] = df["open"] / (df["prev_close"] + 1e-8) - 1.0
    df["intraday_ret"] = df["close"] / (df["open"] + 1e-8) - 1.0
    df["high_rel"] = df["high"] / (df["prev_close"] + 1e-8) - 1.0
    df["low_rel"] = df["low"] / (df["prev_close"] + 1e-8) - 1.0

    # 成交量和成交额相对各自20日均值，消除股票规模及年代数量级差异。
    df["volume_ma20"] = grouped["volume"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["money_ma20"] = grouped["money"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["volume_rel20"] = df["volume"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["money_rel20"] = df["money"] / (df["money_ma20"] + 1e-8) - 1.0

    # -------------------------------------------------------------------------
    # 收益率与动量特征
    # -------------------------------------------------------------------------
    df["pct_chg"] = grouped["close"].pct_change()
    df["mom_5"] = grouped["close"].pct_change(5)
    df["mom_10"] = grouped["close"].pct_change(10)
    df["mom_20"] = grouped["close"].pct_change(20)

    # -------------------------------------------------------------------------
    # 均线距离特征
    # -------------------------------------------------------------------------
    df["ma5"] = grouped["close"].transform(lambda x: x.rolling(5).mean())
    df["ma10"] = grouped["close"].transform(lambda x: x.rolling(10).mean())
    df["ma20"] = grouped["close"].transform(lambda x: x.rolling(20).mean())
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    # -------------------------------------------------------------------------
    # 波动率与量价趋势特征
    # -------------------------------------------------------------------------
    df["vol_ma5"] = grouped["volume"].transform(lambda x: x.rolling(5).mean())
    df["vol_ma5_rel20"] = df["vol_ma5"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["std_5"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    # RSI 14
    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = grouped["close"].transform(lambda x: get_rsi(x, 14))

    # MACD柱状图：除以收盘价，转换成跨股票可比的相对尺度。
    df["ema12"] = grouped["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = grouped["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument", sort=False)["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist_rel"] = ((df["dif"] - df["dea"]) * 2) / (
        df["close"].abs() + 1e-8
    )

    # WR 14
    df["high_14"] = grouped["high"].transform(lambda x: x.rolling(14).max())
    df["low_14"] = grouped["low"].transform(lambda x: x.rolling(14).min())
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    # 对除零或异常行情产生的无穷值统一转为缺失值，由后续dropna剔除。
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    feat_cols = FEATURE_COLUMNS.copy()
    df.drop(
        columns=[
            "prev_close", "volume_ma20", "money_ma20",
            "ma5", "ma10", "ma20", "vol_ma5",
            "ema12", "ema26", "dif", "dea", "high_14", "low_14",
        ],
        errors="ignore",
        inplace=True,
    )
    return df, feat_cols


# =============================================================================
# 5. 模型训练：未来5日累计收益 + 原始/生成样本联合训练
# =============================================================================


def _calculate_daily_rank_ic(
    predictions,
    targets,
    sample_dates,
    sample_instruments,
):
    """
    按样本日期计算验证集横截面RankIC。

    RankIC使用预测值排名与未来5日累计收益排名之间的Pearson相关系数，
    等价于Spearman秩相关。仅保留股票数不少于MIN_RANKIC_STOCKS，
    且预测值和标签均存在横截面差异的日期。
    """
    validation_df = pd.DataFrame(
        {
            "date": pd.to_datetime(sample_dates).normalize(),
            "instrument": np.asarray(sample_instruments, dtype=str),
            "prediction": np.asarray(predictions, dtype=np.float64),
            "target": np.asarray(targets, dtype=np.float64),
        }
    )
    validation_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    validation_df.dropna(
        subset=["date", "instrument", "prediction", "target"],
        inplace=True,
    )

    daily_rank_ic = []
    for date, section in validation_df.groupby("date", sort=True):
        if len(section) < MIN_RANKIC_STOCKS:
            continue
        if (
            section["prediction"].nunique(dropna=True) < 2
            or section["target"].nunique(dropna=True) < 2
        ):
            continue

        prediction_rank = section["prediction"].rank(
            method="average",
            pct=True,
        )
        target_rank = section["target"].rank(
            method="average",
            pct=True,
        )
        rank_ic = prediction_rank.corr(target_rank)
        if pd.notna(rank_ic) and np.isfinite(rank_ic):
            daily_rank_ic.append((date, float(rank_ic), len(section)))

    if not daily_rank_ic:
        return {
            "mean": np.nan,
            "std": np.nan,
            "ir": np.nan,
            "positive_rate": np.nan,
            "days": 0,
        }

    rank_ic_values = np.asarray(
        [item[1] for item in daily_rank_ic],
        dtype=np.float64,
    )
    mean_rank_ic = float(rank_ic_values.mean())
    std_rank_ic = (
        float(rank_ic_values.std(ddof=1))
        if len(rank_ic_values) > 1
        else np.nan
    )
    rank_ic_ir = (
        mean_rank_ic / std_rank_ic
        if np.isfinite(std_rank_ic) and std_rank_ic > 1e-12
        else np.nan
    )

    return {
        "mean": mean_rank_ic,
        "std": std_rank_ic,
        "ir": rank_ic_ir,
        "positive_rate": float((rank_ic_values > 0).mean()),
        "days": int(len(rank_ic_values)),
    }


def _evaluate_gru_on_validation(
    model,
    validation_loader,
    validation_dates,
    validation_instruments,
    criterion,
):
    """在原始标准化验证样本上计算MSE和逐日横截面RankIC。"""
    model.eval()
    predictions = []
    targets = []
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for batch_x, batch_y in validation_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)

            batch_prediction = model(batch_x)
            batch_loss = criterion(batch_prediction, batch_y)

            batch_size = int(batch_x.size(0))
            total_loss += float(batch_loss.item()) * batch_size
            total_count += batch_size

            predictions.append(
                batch_prediction.detach().cpu().numpy().reshape(-1)
            )
            targets.append(batch_y.detach().cpu().numpy().reshape(-1))

    if not predictions:
        raise RuntimeError("验证集没有可用于评估的样本")

    predictions = np.concatenate(predictions)
    targets = np.concatenate(targets)

    rank_ic_metrics = _calculate_daily_rank_ic(
        predictions,
        targets,
        validation_dates,
        validation_instruments,
    )
    validation_mse = total_loss / max(total_count, 1)
    return validation_mse, rank_ic_metrics


def train_and_save_models(context, train_stocks):
    start_fetch = (
        pd.to_datetime(TRAIN_START) - pd.Timedelta(days=40)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"开始训练：股票数={len(train_stocks)}，"
        f"行情区间={start_fetch}~{TRAIN_END}"
    )

    price_df = _query_daily_bars(
        train_stocks,
        start_fetch,
        TRAIN_END,
    )
    if price_df.empty:
        raise RuntimeError("训练行情为空")

    # 对应原版 get_price(..., fq='pre')。
    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df, feat_cols = calculate_all_factors(price_df)

    context.feat_dim = len(feat_cols)

    # 标签：以当前行收盘价为起点，到未来第5个交易日收盘价的累计收益率。
    # target[t] = close[t+5] / close[t] - 1
    grouped_close = price_df.groupby("instrument", sort=False)["close"]
    price_df["target"] = grouped_close.shift(-5) / price_df["close"] - 1.0

    price_df = price_df[
        (price_df["date"] >= pd.Timestamp(TRAIN_START))
        & (price_df["date"] <= pd.Timestamp(TRAIN_END))
    ]
    price_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    price_df.dropna(subset=feat_cols + ["target"], inplace=True)

    x_list = []
    y_list = []
    sample_dates = []
    sample_instruments = []

    for instrument, group in price_df.groupby("instrument", sort=False):
        group = group.sort_values("date").reset_index(drop=True)
        if len(group) < SEQ_LEN:
            continue

        feats = group[feat_cols].to_numpy(dtype=np.float32)
        targets = group["target"].to_numpy(dtype=np.float32)
        dates = group["date"].to_numpy(dtype="datetime64[ns]")

        # 窗口最后一日end_idx直接匹配同一日的未来5日累计收益标签。
        for end_idx in range(SEQ_LEN - 1, len(group)):
            start_idx = end_idx - SEQ_LEN + 1
            x_list.append(feats[start_idx: end_idx + 1])
            y_list.append(targets[end_idx])
            sample_dates.append(dates[end_idx])
            sample_instruments.append(str(instrument))

    if not x_list:
        raise RuntimeError("未构造出有效训练样本")

    x_array = np.asarray(x_list, dtype=np.float32)
    y_array = np.asarray(y_list, dtype=np.float32).reshape(-1, 1)
    sample_dates_array = np.asarray(
        sample_dates,
        dtype="datetime64[ns]",
    )
    sample_instruments_array = np.asarray(
        sample_instruments,
        dtype=str,
    )

    del x_list, y_list, sample_dates, sample_instruments
    gc.collect()

    # -------------------------------------------------------------------------
    # 严格时间划分：按样本窗口结束日期，而不是随机划分。
    # 最后VALIDATION_RATIO的日期作为验证期；验证期之前剔除
    # PURGE_TRADING_DAYS个样本日期，避免5日标签跨越边界。
    # -------------------------------------------------------------------------
    unique_sample_dates = np.sort(np.unique(sample_dates_array))
    if len(unique_sample_dates) < MIN_VALIDATION_DATES * 2:
        raise RuntimeError(
            f"有效样本日期仅{len(unique_sample_dates)}个，"
            "不足以构造严格时间训练集和验证集"
        )

    validation_date_count = max(
        MIN_VALIDATION_DATES,
        int(np.ceil(len(unique_sample_dates) * VALIDATION_RATIO)),
    )
    validation_date_count = min(
        validation_date_count,
        len(unique_sample_dates) - MIN_VALIDATION_DATES,
    )
    validation_start_index = len(unique_sample_dates) - validation_date_count
    train_end_index = validation_start_index - PURGE_TRADING_DAYS - 1

    if train_end_index < MIN_VALIDATION_DATES - 1:
        raise RuntimeError(
            "隔离期后训练日期不足，请缩短验证比例或隔离期"
        )

    train_end_date = unique_sample_dates[train_end_index]
    validation_start_date = unique_sample_dates[validation_start_index]

    train_mask = sample_dates_array <= train_end_date
    validation_mask = sample_dates_array >= validation_start_date

    if not train_mask.any() or not validation_mask.any():
        raise RuntimeError("严格时间划分后训练集或验证集为空")

    x_train = torch.from_numpy(x_array[train_mask].copy())
    y_train = torch.from_numpy(y_array[train_mask].copy())
    x_validation = torch.from_numpy(x_array[validation_mask].copy())
    y_validation = torch.from_numpy(y_array[validation_mask].copy())

    validation_dates = sample_dates_array[validation_mask].copy()
    validation_instruments = sample_instruments_array[validation_mask].copy()

    purged_dates = unique_sample_dates[
        train_end_index + 1:validation_start_index
    ]

    context.logger.info(
        "严格时间划分完成："
        f"训练期截至={pd.Timestamp(train_end_date).date()}，"
        f"验证期开始={pd.Timestamp(validation_start_date).date()}，"
        f"隔离交易日={len(purged_dates)}，"
        f"训练样本={len(x_train):,}，"
        f"验证样本={len(x_validation):,}，"
        f"验证日期={len(np.unique(validation_dates))}"
    )

    del (
        x_array,
        y_array,
        sample_dates_array,
        sample_instruments_array,
        unique_sample_dates,
        train_mask,
        validation_mask,
        purged_dates,
    )
    gc.collect()

    # 标准化参数只使用训练集，验证集不参与任何统计量估计。
    x_flat = x_train.view(-1, context.feat_dim)
    context.train_mean = x_flat.mean(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1)
    context.train_std = x_flat.std(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1) + 1e-8

    x_train = (x_train - context.train_mean) / context.train_std
    x_validation = (
        x_validation - context.train_mean
    ) / context.train_std

    gan_train_loader = DataLoader(
        TensorDataset(x_train, y_train),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )
    validation_loader = DataLoader(
        TensorDataset(x_validation, y_validation),
        batch_size=BATCH_SIZE,
        shuffle=False,
    )

    # GAN仅使用训练集，验证集不参与生成器或判别器训练。
    generator = Generator(context.feat_dim, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        context.feat_dim,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion_gan = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()
        epoch_loss_g = 0.0
        epoch_loss_d = 0.0
        epoch_samples = 0

        for real_data, _ in gan_train_loader:
            real_data = real_data.to(DEVICE)
            batch_size = int(real_data.size(0))
            real_label = torch.ones(batch_size, device=DEVICE)
            fake_label = torch.zeros(batch_size, device=DEVICE)

            discriminator.zero_grad()
            loss_d = (
                criterion_gan(discriminator(real_data), real_label)
                + criterion_gan(
                    discriminator(generator(real_data).detach()),
                    fake_label,
                )
            )
            loss_d.backward()
            opt_d.step()

            generator.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion_gan(discriminator(fake_data), real_label)
                + LAMBDA_REC
                * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

            epoch_loss_d += float(loss_d.item()) * batch_size
            epoch_loss_g += float(loss_g.item()) * batch_size
            epoch_samples += batch_size

        context.logger.info(
            f"GAN训练完成：epoch={epoch + 1}/{GAN_EPOCHS}，"
            f"loss_D={epoch_loss_d / max(epoch_samples, 1):.6f}，"
            f"loss_G={epoch_loss_g / max(epoch_samples, 1):.6f}"
        )

    # 只为训练集生成增强样本；验证集始终使用真实原始标准化样本。
    generator.eval()
    with torch.no_grad():
        x_enhanced = torch.cat(
            [
                generator(
                    x_train[i: i + BATCH_SIZE].to(DEVICE)
                ).cpu()
                for i in range(0, x_train.size(0), BATCH_SIZE)
            ],
            dim=0,
        )

    original_dataset = TensorDataset(x_train, y_train)
    enhanced_dataset = TensorDataset(x_enhanced, y_train)
    joint_dataset = ConcatDataset(
        [original_dataset, enhanced_dataset]
    )
    joint_loader = DataLoader(
        joint_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
    )

    del x_flat, price_df
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # -------------------------------------------------------------------------
    # GRU训练：每轮后在严格时间验证集上计算逐日横截面RankIC。
    # 保存平均RankIC最高的模型，连续EARLY_STOP_PATIENCE轮未改善则停止。
    # -------------------------------------------------------------------------
    context.gru_model = StockGRU(
        context.feat_dim,
        64,
        2,
    ).to(DEVICE)
    optimizer_gru = optim.Adam(
        context.gru_model.parameters(),
        lr=0.001,
    )
    criterion_gru = nn.MSELoss()

    best_rank_ic = -np.inf
    best_model_state = None
    best_epoch = 0
    no_improvement_epochs = 0

    for epoch in range(GRU_MAX_EPOCHS):
        context.gru_model.train()
        total_train_loss = 0.0
        total_train_count = 0

        for batch_x, batch_y in joint_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)

            optimizer_gru.zero_grad()
            prediction = context.gru_model(batch_x)
            loss = criterion_gru(prediction, batch_y)
            loss.backward()
            optimizer_gru.step()

            batch_size = int(batch_x.size(0))
            total_train_loss += float(loss.item()) * batch_size
            total_train_count += batch_size

        train_mse = total_train_loss / max(total_train_count, 1)
        validation_mse, rank_ic_metrics = _evaluate_gru_on_validation(
            context.gru_model,
            validation_loader,
            validation_dates,
            validation_instruments,
            criterion_gru,
        )
        mean_rank_ic = rank_ic_metrics["mean"]

        improved = (
            np.isfinite(mean_rank_ic)
            and mean_rank_ic > best_rank_ic + RANKIC_MIN_DELTA
        )

        if improved:
            best_rank_ic = float(mean_rank_ic)
            best_epoch = epoch + 1
            no_improvement_epochs = 0
            best_model_state = {
                key: value.detach().cpu().clone()
                for key, value
                in context.gru_model.state_dict().items()
            }
        else:
            no_improvement_epochs += 1

        context.logger.info(
            f"GRU epoch={epoch + 1}/{GRU_MAX_EPOCHS}，"
            f"train_MSE={train_mse:.8f}，"
            f"val_MSE={validation_mse:.8f}，"
            f"val_RankIC={mean_rank_ic:.6f}，"
            f"RankICIR={rank_ic_metrics['ir']:.4f}，"
            f"RankIC胜率={rank_ic_metrics['positive_rate']:.4f}，"
            f"有效截面={rank_ic_metrics['days']}，"
            f"最佳epoch={best_epoch}，"
            f"未改善轮数={no_improvement_epochs}"
        )

        if no_improvement_epochs >= EARLY_STOP_PATIENCE:
            context.logger.info(
                f"触发RankIC早停：连续{EARLY_STOP_PATIENCE}轮未改善"
            )
            break

    if best_model_state is None:
        raise RuntimeError(
            "验证集未计算出有效RankIC，无法选择最佳GRU模型"
        )

    context.gru_model.load_state_dict(best_model_state)
    context.gru_model.to(DEVICE)
    context.gru_model.eval()
    context.model_ready = True
    context.best_gru_epoch = best_epoch
    context.best_validation_rank_ic = best_rank_ic
    context.validation_start_date = pd.Timestamp(
        validation_start_date
    ).strftime("%Y-%m-%d")

    del (
        x_train,
        x_validation,
        x_enhanced,
        y_train,
        y_validation,
        original_dataset,
        enhanced_dataset,
        joint_dataset,
        generator,
        discriminator,
        best_model_state,
        validation_dates,
        validation_instruments,
    )
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.logger.info(
        "GAN + GRU模型已就绪："
        "目标=未来5日累计收益，"
        "GRU训练集=原始样本+生成样本，"
        f"最佳epoch={best_epoch}，"
        f"最佳验证RankIC={best_rank_ic:.6f}"
    )


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None

    # 权重数据在bigtrader.run之前只读取一次，同时用于限定回测股票池。
    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.logger.info(
        f"已加载沪深300权重缓存：{len(context.index_weights_df):,}行，"
        f"股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    # 训练股票池：训练结束日的沪深300成分股。
    train_weights = _weights_asof(context.index_weights_df, TRAIN_END)
    train_stocks = train_weights["instrument"].tolist()
    if not train_stocks:
        raise RuntimeError(f"无法取得{TRAIN_END}沪深300成分股")

    train_and_save_models(context, train_stocks)

    # 一次性读取回测期可能用到的全部日线原始数据；每周预测时仍按原版逻辑
    # 截取当期股票最近100个完整日线，并重新做末端前复权与因子计算。
    backtest_sections = context.index_weights_df[
        (context.index_weights_df["date"] >= pd.Timestamp(context.start_date) - pd.Timedelta(days=180))
        & (context.index_weights_df["date"] <= pd.Timestamp(context.end_date))
    ]
    universe = sorted(
        backtest_sections["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300股票并集为空")

    daily_start = (
        pd.Timestamp(context.start_date) - pd.Timedelta(days=220)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"读取回测期日线缓存：{daily_start}~{context.end_date}，"
        f"股票并集={len(universe)}"
    )
    daily_df = _query_daily_bars(
        universe,
        daily_start,
        context.end_date,
    )
    if daily_df.empty:
        raise RuntimeError("回测期日线缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }

    del daily_df, backtest_sections
    gc.collect()


def _build_realtime_factor_data(context, stocks, current_date):
    """
    复刻原版：
        get_price(stocks, end_date=current_date,
                  count=SEQ_LEN+60, frequency='daily', fq='pre')

    日频回测在周五调仓，只使用current_date之前已经完整结束的日线。
    因此信号在周五交易前已经可得，不使用周五收盘后信息。
    """
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue

        # 原版count=100；从完整日线中取最近100条。
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if section.empty:
            continue
        pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df, feat_cols = calculate_all_factors(realtime_df)
    realtime_df.dropna(inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(
        context,
        stocks,
        current_date,
    )
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    # 批量推理只减少Python循环和模型调用次数，不改变每只股票的输入、
    # 标准化方式、模型结构或预测结果含义。预测值对应未来5日累计收益，实时推理仍不经过Generator。
    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )

    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 回测静态数据准备
# =============================================================================

def _prepare_backtest_static_data():
    """
    在启动BigTrader前读取一次指数权重，并据此限定回测股票池。

    这样日频引擎只加载回测期沪深300历史成分股并集，避免在未指定
    instruments时准备全市场行情。
    """
    weight_start = (
        min(pd.Timestamp(TRAIN_END), pd.Timestamp(BACKTEST_START))
        - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        using_official = False

    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    return weights, using_official, universe


(
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
) = _prepare_backtest_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

```

## 改进版模型3

本版本在原有 GAN-GRU 时序模型基础上，进一步加入了股票之间的横截面关系。

1. 扩展模型输入

模型输入由原来的股票自身时序特征，扩展为：

股票过去40日的量价时序特征；
每日横截面分位排名特征；
沪深300市场状态特征；
股票所属行业的状态特征。

这样模型不仅能观察单只股票自身的变化，也能判断其相对于市场和行业的位置。

2. 按日期组织训练样本

训练时不再将不同日期、不同股票随机混合，而是将同一天的全部有效股票组成一个完整横截面。

GRU仍然共享参数并分别提取每只股票的时序信息，但损失函数会统一比较同一日期内所有股票的预测结果。

3. 加入横截面排序损失

模型损失由单独的收益预测误差改为：

[
总损失 = MSE损失 + 横截面排序损失
]

MSE损失用于预测未来5日累计收益，排序损失用于提高同一日期内股票预测分数与实际收益排名的一致性。

4. 保留GAN数据增强

GRU继续使用原始样本和GAN生成样本共同训练。GAN主要增强股票自身的时序特征，不改变市场、行业及横截面排名信息。

5. 使用RankIC早停

每轮训练结束后，模型会在严格划分的时间验证集上计算每日横截面RankIC。

最终使用验证集平均RankIC最高的模型，并在RankIC连续多轮没有改善时提前停止训练，以降低过拟合风险。


```python
# -*- coding: utf-8 -*-
"""
GAN + 共享GRU 沪深300横截面指数增强策略
BigQuant DAI + BigTrader 原生日频回测版

本版在“未来5日累计收益 + GAN增强 + 时间验证集 + RankIC早停”版本上继续改进：
1. 股票自身20维量价时序特征；
2. 对20个股票特征增加每日沪深300横截面分位排名；
3. 增加沪深300市场状态特征和申万一级行业状态特征；
4. 共享GRU仍逐只股票编码40日序列，但训练时按日期组成完整股票截面；
5. GRU损失改为：原始/生成样本MSE + 横截面排序损失；
6. 仍使用严格时间验证集，并以验证集逐日RankIC早停。

BigQuant官方数据接口：
- cn_stock_bar1d：股票日线；
- cn_stock_index_bar1d：沪深300指数日线；
- cn_stock_factors_base.sw2021_level1：申万2021一级行业代码；
- cn_stock_index_weight：沪深300官方权重；
- bigtrader.run(..., frequency=bigtrader.Frequency.DAILY)。
"""

import gc
import random
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# =============================================================================
# 1. 参数
# =============================================================================

INDEX_CODE = "000300.SH"
INDUSTRY_STANDARD = "sw2021"

TRAIN_START = "2023-01-01"
TRAIN_END = "2024-05-01"
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2024-08-29"
CAPITAL_BASE = 1_000_000

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5
GAN_EPOCHS = 4

GRU_MAX_EPOCHS = 20
EARLY_STOP_PATIENCE = 4
RANKIC_MIN_DELTA = 1e-4
RANK_LOSS_WEIGHT = 0.01

VALIDATION_RATIO = 0.20
PURGE_TRADING_DAYS = 5
MIN_VALIDATION_DATES = 20
MIN_RANKIC_STOCKS = 30
MIN_TRAIN_CROSS_SECTION_STOCKS = 30

BATCH_SIZE = 1024
BIAS = 0.008
DAILY_LOOKBACK_COUNT = SEQ_LEN + 60
TRAIN_FETCH_BUFFER_DAYS = 180
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

BASE_FEATURE_COLUMNS = [
    "open_gap", "intraday_ret", "high_rel", "low_rel",
    "volume_rel20", "money_rel20",
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5_rel20",
    "std_5", "std_20", "rsi_14", "macd_hist_rel", "wr_14",
    "volume_ratio",
]

CROSS_SECTIONAL_RANK_COLUMNS = [
    f"{column}_cs_rank" for column in BASE_FEATURE_COLUMNS
]

MARKET_FEATURE_COLUMNS = [
    "mkt_ret_1", "mkt_mom_5", "mkt_mom_20", "mkt_vol_20",
    "mkt_volume_rel20", "mkt_amount_rel20",
]

INDUSTRY_FEATURE_COLUMNS = [
    "ind_ret_1", "ind_mom_5", "ind_mom_20", "ind_vol_20",
    "ind_breadth", "ind_money_rel20",
]

FEATURE_COLUMNS = (
    BASE_FEATURE_COLUMNS
    + CROSS_SECTIONAL_RANK_COLUMNS
    + MARKET_FEATURE_COLUMNS
    + INDUSTRY_FEATURE_COLUMNS
)
BASE_FEATURE_DIM = len(BASE_FEATURE_COLUMNS)


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)




def _query_industry_codes(stocks, start_date, end_date):
    """从官方基础因子表读取申万2021一级行业代码。"""
    if not stocks:
        return pd.DataFrame(columns=["date", "instrument", "industry_code"])

    sql = """
    SELECT
        date,
        instrument,
        sw2021_level1 AS industry_code
    FROM cn_stock_factors_base
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """
    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return pd.DataFrame(columns=["date", "instrument", "industry_code"])

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["industry_code"] = df["industry_code"].astype("string")
    return (
        df[["date", "instrument", "industry_code"]]
        .drop_duplicates(["date", "instrument"], keep="last")
        .sort_values(["instrument", "date"])
        .reset_index(drop=True)
    )


def _attach_industry_codes(price_df, industry_df):
    """按日期连接行业代码；仅向前填充，避免使用未来行业信息。"""
    if price_df.empty:
        return price_df

    result = price_df.copy()
    if industry_df is None or industry_df.empty:
        result["industry_code"] = "UNKNOWN"
        return result

    result = result.merge(
        industry_df,
        on=["date", "instrument"],
        how="left",
        validate="one_to_one",
    )
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)
    result["industry_code"] = (
        result.groupby("instrument", sort=False)["industry_code"]
        .ffill()
        .fillna("UNKNOWN")
        .astype(str)
    )
    return result


def _query_index_bars(start_date, end_date):
    """读取沪深300指数日线，用于构造市场状态。"""
    sql = """
    SELECT
        date,
        close,
        volume,
        amount
    FROM cn_stock_index_bar1d
    WHERE instrument = $index_code
    ORDER BY date
    """
    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    for column in ["close", "volume", "amount"]:
        df[column] = pd.to_numeric(df[column], errors="coerce")
    return (
        df.sort_values("date")
        .drop_duplicates("date", keep="last")
        .reset_index(drop=True)
    )


def _calculate_market_state_features(index_df):
    """构造6个沪深300市场状态变量。"""
    if index_df is None or index_df.empty:
        return pd.DataFrame(columns=["date"] + MARKET_FEATURE_COLUMNS)

    result = index_df.copy().sort_values("date").reset_index(drop=True)
    result["mkt_ret_1"] = result["close"].pct_change()
    result["mkt_mom_5"] = result["close"].pct_change(5)
    result["mkt_mom_20"] = result["close"].pct_change(20)
    result["mkt_vol_20"] = result["mkt_ret_1"].rolling(20).std()
    result["mkt_volume_rel20"] = (
        result["volume"] / (result["volume"].rolling(20).mean() + 1e-8) - 1.0
    )
    result["mkt_amount_rel20"] = (
        result["amount"] / (result["amount"].rolling(20).mean() + 1e-8) - 1.0
    )
    result.replace([np.inf, -np.inf], np.nan, inplace=True)
    return result[["date"] + MARKET_FEATURE_COLUMNS]

# =============================================================================
# 4. 股票、横截面、市场与行业特征
# =============================================================================

def calculate_base_factors(df):
    """构造20维股票自身量价时序特征。"""
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)
    grouped = df.groupby("instrument", sort=False)

    df["prev_close"] = grouped["close"].shift(1)
    df["open_gap"] = df["open"] / (df["prev_close"] + 1e-8) - 1.0
    df["intraday_ret"] = df["close"] / (df["open"] + 1e-8) - 1.0
    df["high_rel"] = df["high"] / (df["prev_close"] + 1e-8) - 1.0
    df["low_rel"] = df["low"] / (df["prev_close"] + 1e-8) - 1.0

    df["volume_ma20"] = grouped["volume"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["money_ma20"] = grouped["money"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["volume_rel20"] = df["volume"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["money_rel20"] = df["money"] / (df["money_ma20"] + 1e-8) - 1.0

    df["pct_chg"] = grouped["close"].pct_change()
    df["mom_5"] = grouped["close"].pct_change(5)
    df["mom_10"] = grouped["close"].pct_change(10)
    df["mom_20"] = grouped["close"].pct_change(20)

    df["ma5"] = grouped["close"].transform(lambda x: x.rolling(5).mean())
    df["ma10"] = grouped["close"].transform(lambda x: x.rolling(10).mean())
    df["ma20"] = grouped["close"].transform(lambda x: x.rolling(20).mean())
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    df["vol_ma5"] = grouped["volume"].transform(lambda x: x.rolling(5).mean())
    df["vol_ma5_rel20"] = df["vol_ma5"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["std_5"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = grouped["close"].transform(lambda x: get_rsi(x, 14))
    df["ema12"] = grouped["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = grouped["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument", sort=False)["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist_rel"] = ((df["dif"] - df["dea"]) * 2) / (
        df["close"].abs() + 1e-8
    )

    df["high_14"] = grouped["high"].transform(lambda x: x.rolling(14).max())
    df["low_14"] = grouped["low"].transform(lambda x: x.rolling(14).min())
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.drop(
        columns=[
            "prev_close", "volume_ma20", "money_ma20",
            "ma5", "ma10", "ma20", "vol_ma5",
            "ema12", "ema26", "dif", "dea", "high_14", "low_14",
        ],
        errors="ignore",
        inplace=True,
    )
    return df


def _add_contextual_features(stock_df, market_state_df):
    """
    增加：
    - 20个每日横截面分位排名，映射到[-1, 1]；
    - 6个沪深300市场状态；
    - 6个申万一级行业状态。
    """
    if stock_df.empty:
        return stock_df, FEATURE_COLUMNS.copy()

    df = stock_df.copy().sort_values(["date", "instrument"]).reset_index(drop=True)
    df["industry_code"] = df["industry_code"].fillna("UNKNOWN").astype(str)

    for column, rank_column in zip(
        BASE_FEATURE_COLUMNS,
        CROSS_SECTIONAL_RANK_COLUMNS,
    ):
        percentile = df.groupby("date", sort=False)[column].rank(
            method="average",
            pct=True,
        )
        df[rank_column] = percentile * 2.0 - 1.0

    industry_daily = (
        df.groupby(["industry_code", "date"], sort=True)
        .agg(
            ind_ret_1=("pct_chg", "mean"),
            ind_breadth=("pct_chg", lambda x: float((x > 0).mean())),
            ind_money_rel20=("money_rel20", "mean"),
        )
        .reset_index()
        .sort_values(["industry_code", "date"])
        .reset_index(drop=True)
    )

    safe_industry_return = industry_daily["ind_ret_1"].clip(lower=-0.95)
    industry_daily["industry_nav"] = (
        1.0 + safe_industry_return.fillna(0.0)
    ).groupby(industry_daily["industry_code"]).cumprod()

    industry_group = industry_daily.groupby("industry_code", sort=False)
    industry_daily["ind_mom_5"] = industry_group["industry_nav"].pct_change(5)
    industry_daily["ind_mom_20"] = industry_group["industry_nav"].pct_change(20)
    industry_daily["ind_vol_20"] = industry_group["ind_ret_1"].transform(
        lambda x: x.rolling(20).std()
    )
    industry_daily.drop(columns=["industry_nav"], inplace=True)

    df = df.merge(
        industry_daily,
        on=["industry_code", "date"],
        how="left",
        validate="many_to_one",
    )
    df = df.merge(
        market_state_df,
        on="date",
        how="left",
        validate="many_to_one",
    )
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    return df, FEATURE_COLUMNS.copy()


# =============================================================================
# 5. 按日期完整横截面训练：MSE + 排序损失 + RankIC早停
# =============================================================================

def _calculate_daily_rank_ic(predictions, targets, sample_dates, sample_instruments):
    validation_df = pd.DataFrame(
        {
            "date": pd.to_datetime(sample_dates).normalize(),
            "instrument": np.asarray(sample_instruments, dtype=str),
            "prediction": np.asarray(predictions, dtype=np.float64),
            "target": np.asarray(targets, dtype=np.float64),
        }
    )
    validation_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    validation_df.dropna(
        subset=["date", "instrument", "prediction", "target"],
        inplace=True,
    )

    values = []
    for date, section in validation_df.groupby("date", sort=True):
        if len(section) < MIN_RANKIC_STOCKS:
            continue
        if section["prediction"].nunique() < 2 or section["target"].nunique() < 2:
            continue
        pred_rank = section["prediction"].rank(method="average", pct=True)
        target_rank = section["target"].rank(method="average", pct=True)
        rank_ic = pred_rank.corr(target_rank)
        if pd.notna(rank_ic) and np.isfinite(rank_ic):
            values.append(float(rank_ic))

    if not values:
        return {
            "mean": np.nan,
            "std": np.nan,
            "ir": np.nan,
            "positive_rate": np.nan,
            "days": 0,
        }

    array = np.asarray(values, dtype=np.float64)
    mean = float(array.mean())
    std = float(array.std(ddof=1)) if len(array) > 1 else np.nan
    return {
        "mean": mean,
        "std": std,
        "ir": mean / std if np.isfinite(std) and std > 1e-12 else np.nan,
        "positive_rate": float((array > 0).mean()),
        "days": int(len(array)),
    }


def _build_date_groups(dates, targets, min_stocks):
    """为每个日期建立完整横截面索引和固定的真实收益分位排名。"""
    dates = np.asarray(dates, dtype="datetime64[ns]")
    targets = np.asarray(targets, dtype=np.float32).reshape(-1)
    groups = []

    for date in np.sort(np.unique(dates)):
        indices = np.flatnonzero(dates == date)
        valid = np.isfinite(targets[indices])
        indices = indices[valid]
        if len(indices) < min_stocks:
            continue

        section_targets = targets[indices]
        if np.unique(section_targets).size < 2:
            continue

        target_rank = (
            pd.Series(section_targets)
            .rank(method="average", pct=True)
            .to_numpy(dtype=np.float32)
            .reshape(-1, 1)
        )
        groups.append(
            {
                "date": date,
                "indices": torch.from_numpy(indices.astype(np.int64)),
                "target_rank": torch.from_numpy(target_rank),
            }
        )
    return groups


def _cross_section_rank_loss(prediction, target_rank):
    """1减去预测值与真实收益分位排名的Pearson相关系数。"""
    pred = prediction.reshape(-1)
    rank = target_rank.reshape(-1)
    pred = pred - pred.mean()
    rank = rank - rank.mean()
    denominator = torch.sqrt(torch.sum(pred * pred) + 1e-12) * torch.sqrt(
        torch.sum(rank * rank) + 1e-12
    )
    correlation = torch.sum(pred * rank) / denominator
    return 1.0 - correlation


def _evaluate_validation_by_date(
    model,
    x_validation,
    y_validation,
    validation_groups,
    validation_dates,
    validation_instruments,
    criterion,
):
    model.eval()
    predictions = np.full(len(x_validation), np.nan, dtype=np.float32)
    targets = y_validation.numpy().reshape(-1)
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for group in validation_groups:
            indices = group["indices"]
            batch_x = x_validation.index_select(0, indices).to(DEVICE)
            batch_y = y_validation.index_select(0, indices).to(DEVICE)
            batch_prediction = model(batch_x)
            batch_loss = criterion(batch_prediction, batch_y)

            count = int(len(indices))
            total_loss += float(batch_loss.item()) * count
            total_count += count
            predictions[indices.numpy()] = (
                batch_prediction.detach().cpu().numpy().reshape(-1)
            )

    valid_mask = np.isfinite(predictions)
    if not valid_mask.any():
        raise RuntimeError("验证集没有可用于完整横截面评估的样本")

    rank_metrics = _calculate_daily_rank_ic(
        predictions[valid_mask],
        targets[valid_mask],
        validation_dates[valid_mask],
        validation_instruments[valid_mask],
    )
    return total_loss / max(total_count, 1), rank_metrics


def train_and_save_models(context, train_stocks):
    start_fetch = (
        pd.Timestamp(TRAIN_START) - pd.Timedelta(days=TRAIN_FETCH_BUFFER_DAYS)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"开始训练：股票数={len(train_stocks)}，"
        f"行情区间={start_fetch}~{TRAIN_END}"
    )

    price_df = _query_daily_bars(train_stocks, start_fetch, TRAIN_END)
    if price_df.empty:
        raise RuntimeError("训练行情为空")

    industry_df = _query_industry_codes(train_stocks, start_fetch, TRAIN_END)
    market_raw = _query_index_bars(start_fetch, TRAIN_END)
    market_state = _calculate_market_state_features(market_raw)
    if market_state.empty:
        raise RuntimeError("沪深300市场状态数据为空")

    price_df = _attach_industry_codes(price_df, industry_df)
    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df = calculate_base_factors(price_df)
    price_df, feat_cols = _add_contextual_features(price_df, market_state)

    context.feat_dim = len(feat_cols)
    grouped_close = price_df.groupby("instrument", sort=False)["close"]
    price_df["target"] = grouped_close.shift(-5) / price_df["close"] - 1.0

    price_df = price_df[
        (price_df["date"] >= pd.Timestamp(TRAIN_START))
        & (price_df["date"] <= pd.Timestamp(TRAIN_END))
    ].copy()
    price_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    price_df.dropna(subset=feat_cols + ["target"], inplace=True)

    x_list, y_list, date_list, instrument_list = [], [], [], []
    for instrument, group in price_df.groupby("instrument", sort=False):
        group = group.sort_values("date").reset_index(drop=True)
        if len(group) < SEQ_LEN:
            continue
        feats = group[feat_cols].to_numpy(dtype=np.float32)
        targets = group["target"].to_numpy(dtype=np.float32)
        dates = group["date"].to_numpy(dtype="datetime64[ns]")

        for end_idx in range(SEQ_LEN - 1, len(group)):
            start_idx = end_idx - SEQ_LEN + 1
            x_list.append(feats[start_idx:end_idx + 1])
            y_list.append(targets[end_idx])
            date_list.append(dates[end_idx])
            instrument_list.append(str(instrument))

    if not x_list:
        raise RuntimeError("未构造出有效训练样本")

    x_array = np.asarray(x_list, dtype=np.float32)
    y_array = np.asarray(y_list, dtype=np.float32).reshape(-1, 1)
    dates_array = np.asarray(date_list, dtype="datetime64[ns]")
    instruments_array = np.asarray(instrument_list, dtype=str)
    del x_list, y_list, date_list, instrument_list, price_df
    gc.collect()

    unique_dates = np.sort(np.unique(dates_array))
    if len(unique_dates) < MIN_VALIDATION_DATES * 2:
        raise RuntimeError("有效样本日期不足以构造严格时间验证集")

    validation_date_count = max(
        MIN_VALIDATION_DATES,
        int(np.ceil(len(unique_dates) * VALIDATION_RATIO)),
    )
    validation_date_count = min(
        validation_date_count,
        len(unique_dates) - MIN_VALIDATION_DATES,
    )
    validation_start_index = len(unique_dates) - validation_date_count
    train_end_index = validation_start_index - PURGE_TRADING_DAYS - 1
    if train_end_index < MIN_VALIDATION_DATES - 1:
        raise RuntimeError("隔离期后训练日期不足")

    train_end_date = unique_dates[train_end_index]
    validation_start_date = unique_dates[validation_start_index]
    train_mask = dates_array <= train_end_date
    validation_mask = dates_array >= validation_start_date

    x_train = torch.from_numpy(x_array[train_mask].copy())
    y_train = torch.from_numpy(y_array[train_mask].copy())
    train_dates = dates_array[train_mask].copy()
    train_instruments = instruments_array[train_mask].copy()

    x_validation = torch.from_numpy(x_array[validation_mask].copy())
    y_validation = torch.from_numpy(y_array[validation_mask].copy())
    validation_dates = dates_array[validation_mask].copy()
    validation_instruments = instruments_array[validation_mask].copy()

    del x_array, y_array, dates_array, instruments_array, train_mask, validation_mask
    gc.collect()

    x_flat = x_train.view(-1, context.feat_dim)
    context.train_mean = x_flat.mean(dim=0, keepdim=True).view(1, 1, -1)
    context.train_std = x_flat.std(dim=0, keepdim=True).view(1, 1, -1) + 1e-8
    x_train = (x_train - context.train_mean) / context.train_std
    x_validation = (x_validation - context.train_mean) / context.train_std

    train_groups = _build_date_groups(
        train_dates,
        y_train.numpy(),
        MIN_TRAIN_CROSS_SECTION_STOCKS,
    )
    validation_groups = _build_date_groups(
        validation_dates,
        y_validation.numpy(),
        MIN_RANKIC_STOCKS,
    )
    if not train_groups or not validation_groups:
        raise RuntimeError("无法构造有效的训练或验证横截面")

    context.logger.info(
        "严格时间与横截面划分完成："
        f"训练截至={pd.Timestamp(train_end_date).date()}，"
        f"验证开始={pd.Timestamp(validation_start_date).date()}，"
        f"训练截面={len(train_groups)}，验证截面={len(validation_groups)}，"
        f"输入维度={context.feat_dim}"
    )

    # GAN仅增强20维股票自身特征；横截面排名、市场与行业状态保持原值。
    gan_base_train = x_train[:, :, :BASE_FEATURE_DIM].contiguous()
    gan_loader = DataLoader(
        TensorDataset(gan_base_train),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )
    generator = Generator(BASE_FEATURE_DIM, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        BASE_FEATURE_DIM,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion_gan = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()
        loss_g_sum = 0.0
        loss_d_sum = 0.0
        count_sum = 0

        for (real_data,) in gan_loader:
            real_data = real_data.to(DEVICE)
            count = int(real_data.size(0))
            real_label = torch.ones(count, device=DEVICE)
            fake_label = torch.zeros(count, device=DEVICE)

            opt_d.zero_grad()
            fake_detached = generator(real_data).detach()
            loss_d = (
                criterion_gan(discriminator(real_data), real_label)
                + criterion_gan(discriminator(fake_detached), fake_label)
            )
            loss_d.backward()
            opt_d.step()

            opt_g.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion_gan(discriminator(fake_data), real_label)
                + LAMBDA_REC * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

            loss_g_sum += float(loss_g.item()) * count
            loss_d_sum += float(loss_d.item()) * count
            count_sum += count

        context.logger.info(
            f"GAN epoch={epoch + 1}/{GAN_EPOCHS}，"
            f"loss_D={loss_d_sum / max(count_sum, 1):.6f}，"
            f"loss_G={loss_g_sum / max(count_sum, 1):.6f}"
        )

    generator.eval()
    generated_parts = []
    with torch.no_grad():
        for start in range(0, len(gan_base_train), BATCH_SIZE):
            generated_parts.append(
                generator(
                    gan_base_train[start:start + BATCH_SIZE].to(DEVICE)
                ).cpu()
            )
    generated_base = torch.cat(generated_parts, dim=0)
    x_enhanced = x_train.clone()
    x_enhanced[:, :, :BASE_FEATURE_DIM] = generated_base

    del gan_base_train, generated_base, generated_parts, x_flat, market_raw, market_state, industry_df
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.gru_model = StockGRU(context.feat_dim, 64, 2).to(DEVICE)
    optimizer_gru = optim.Adam(context.gru_model.parameters(), lr=0.001)
    criterion_mse = nn.MSELoss()

    best_rank_ic = -np.inf
    best_model_state = None
    best_epoch = 0
    no_improvement_epochs = 0

    for epoch in range(GRU_MAX_EPOCHS):
        context.gru_model.train()
        order = np.random.permutation(len(train_groups))
        mse_sum = 0.0
        rank_loss_sum = 0.0
        total_loss_sum = 0.0
        stock_count = 0

        for group_index in order:
            group = train_groups[group_index]
            indices = group["indices"]
            original_x = x_train.index_select(0, indices).to(DEVICE)
            enhanced_x = x_enhanced.index_select(0, indices).to(DEVICE)
            target_y = y_train.index_select(0, indices).to(DEVICE)
            target_rank = group["target_rank"].to(DEVICE)

            optimizer_gru.zero_grad()
            pred_original = context.gru_model(original_x)
            pred_enhanced = context.gru_model(enhanced_x)

            mse_loss = 0.5 * (
                criterion_mse(pred_original, target_y)
                + criterion_mse(pred_enhanced, target_y)
            )
            rank_loss = _cross_section_rank_loss(pred_original, target_rank)
            total_loss = mse_loss + RANK_LOSS_WEIGHT * rank_loss
            total_loss.backward()
            torch.nn.utils.clip_grad_norm_(context.gru_model.parameters(), 5.0)
            optimizer_gru.step()

            count = int(len(indices))
            mse_sum += float(mse_loss.item()) * count
            rank_loss_sum += float(rank_loss.item()) * count
            total_loss_sum += float(total_loss.item()) * count
            stock_count += count

        validation_mse, rank_metrics = _evaluate_validation_by_date(
            context.gru_model,
            x_validation,
            y_validation,
            validation_groups,
            validation_dates,
            validation_instruments,
            criterion_mse,
        )
        mean_rank_ic = rank_metrics["mean"]
        improved = (
            np.isfinite(mean_rank_ic)
            and mean_rank_ic > best_rank_ic + RANKIC_MIN_DELTA
        )

        if improved:
            best_rank_ic = float(mean_rank_ic)
            best_epoch = epoch + 1
            no_improvement_epochs = 0
            best_model_state = {
                key: value.detach().cpu().clone()
                for key, value in context.gru_model.state_dict().items()
            }
        else:
            no_improvement_epochs += 1

        context.logger.info(
            f"GRU epoch={epoch + 1}/{GRU_MAX_EPOCHS}，"
            f"train_MSE={mse_sum / max(stock_count, 1):.8f}，"
            f"train_rank_loss={rank_loss_sum / max(stock_count, 1):.6f}，"
            f"train_total={total_loss_sum / max(stock_count, 1):.8f}，"
            f"val_MSE={validation_mse:.8f}，"
            f"val_RankIC={mean_rank_ic:.6f}，"
            f"RankICIR={rank_metrics['ir']:.4f}，"
            f"RankIC胜率={rank_metrics['positive_rate']:.4f}，"
            f"有效截面={rank_metrics['days']}，"
            f"最佳epoch={best_epoch}，未改善轮数={no_improvement_epochs}"
        )

        if no_improvement_epochs >= EARLY_STOP_PATIENCE:
            context.logger.info(
                f"触发RankIC早停：连续{EARLY_STOP_PATIENCE}轮未改善"
            )
            break

    if best_model_state is None:
        raise RuntimeError("验证集未得到有效RankIC，无法选择最佳模型")

    context.gru_model.load_state_dict(best_model_state)
    context.gru_model.to(DEVICE)
    context.gru_model.eval()
    context.model_ready = True
    context.best_gru_epoch = best_epoch
    context.best_validation_rank_ic = best_rank_ic

    del (
        x_train, x_validation, x_enhanced, y_train, y_validation,
        train_groups, validation_groups, generator, discriminator,
        best_model_state, train_dates, train_instruments,
        validation_dates, validation_instruments,
    )
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.logger.info(
        "模型已就绪：股票特征20 + 横截面排名20 + 市场状态6 + 行业状态6，"
        f"最佳epoch={best_epoch}，最佳验证RankIC={best_rank_ic:.6f}"
    )


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None

    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.logger.info(
        f"已加载沪深300权重缓存：{len(context.index_weights_df):,}行，"
        f"股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    train_weights = _weights_asof(context.index_weights_df, TRAIN_END)
    train_stocks = train_weights["instrument"].tolist()
    if not train_stocks:
        raise RuntimeError(f"无法取得{TRAIN_END}沪深300成分股")

    train_and_save_models(context, train_stocks)

    backtest_sections = context.index_weights_df[
        (context.index_weights_df["date"] >= pd.Timestamp(context.start_date) - pd.Timedelta(days=180))
        & (context.index_weights_df["date"] <= pd.Timestamp(context.end_date))
    ]
    universe = sorted(
        backtest_sections["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300股票并集为空")

    daily_start = (
        pd.Timestamp(context.start_date) - pd.Timedelta(days=220)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"读取回测期日线与行业缓存：{daily_start}~{context.end_date}，"
        f"股票并集={len(universe)}"
    )
    daily_df = _query_daily_bars(universe, daily_start, context.end_date)
    industry_df = _query_industry_codes(universe, daily_start, context.end_date)
    daily_df = _attach_industry_codes(daily_df, industry_df)
    if daily_df.empty:
        raise RuntimeError("回测期日线缓存为空")

    market_raw = _query_index_bars(daily_start, context.end_date)
    context.market_state_df = _calculate_market_state_features(market_raw)
    if context.market_state_df.empty:
        raise RuntimeError("回测期市场状态缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }

    del daily_df, industry_df, market_raw, backtest_sections
    gc.collect()


def _build_realtime_factor_data(context, stocks, current_date):
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if not section.empty:
            pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df = calculate_base_factors(realtime_df)
    market_section = context.market_state_df[
        context.market_state_df["date"] < cutoff
    ].copy()
    realtime_df, feat_cols = _add_contextual_features(
        realtime_df,
        market_section,
    )
    realtime_df.dropna(subset=feat_cols, inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(context, stocks, current_date)
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        group = group.sort_values("date")
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )
    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 回测静态数据准备
# =============================================================================

def _prepare_backtest_static_data():
    """
    在启动BigTrader前读取一次指数权重，并据此限定回测股票池。

    这样日频引擎只加载回测期沪深300历史成分股并集，避免在未指定
    instruments时准备全市场行情。
    """
    weight_start = (
        min(pd.Timestamp(TRAIN_END), pd.Timestamp(BACKTEST_START))
        - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        using_official = False

    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    return weights, using_official, universe


(
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
) = _prepare_backtest_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

```

## 改进版模型4

```python
# -*- coding: utf-8 -*-
"""
GAN + GRU 沪深300指数增强策略
BigQuant DAI + BigTrader 原生日频回测改进版

本版在原日频移植版本基础上完成三项核心修改：
1. 标签改为窗口末日到未来第5个交易日的累计收益率：
       future_return_5 = close.shift(-5) / close - 1
2. 修正样本索引：每个40日窗口直接匹配窗口最后一日对应的future_return_5；
3. GRU使用“原始标准化样本 + Generator生成样本”共同训练，不再用生成样本替代原始样本；
4. 将直接进入模型的绝对价格、成交量、成交额改为相对特征，并同步消除
   5日均量和MACD柱的绝对尺度；
5. 按样本日期严格划分训练集与验证集，在两者之间留出5个交易日隔离期；
6. GAN仅使用训练集，GRU每轮后计算验证集逐日横截面RankIC，并按RankIC早停；
7. GRU训练批次按日期组织，每次参数更新使用同一交易日的完整股票截面。

保持不变的部分：
- 固定训练期、GAN/GRU网络结构和主要超参数；
- 40日输入窗口；
- 每周五调仓；
- 沪深300官方权重基础上的五组绝对权重偏移规则；
- BigQuant DAI数据接口与BigTrader原生日频回测引擎；
- 实时预测直接输入GRU，不经过Generator。

BigQuant接口：
- from bigquant import bigtrader, dai
- bigtrader.run(..., frequency=bigtrader.Frequency.DAILY)
- cn_stock_bar1d日线数据，amount映射为成交额money
- cn_stock_index_weight指数权重
- context.order_target_percent目标仓位下单
"""

from datetime import datetime
import gc
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# 当前BigQuant DAI环境不提供dai.get_datasource_schema，
# 因此本策略只使用官方文档明确给出的固定表名和字段，不做运行时schema探测。


# =============================================================================
# 1. 参数：与原版保持一致
# =============================================================================

INDEX_CODE = "000300.SH"

TRAIN_START = "2023-01-01"
TRAIN_END = "2024-05-01"

# 原代码未包含回测区间；BigTrader必须显式给出，因此从训练期结束后的下一交易周开始。
# 可只修改这三个回测参数，不影响策略逻辑。
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2026-07-13"
CAPITAL_BASE = 1_000_000

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5

GAN_EPOCHS = 4

# GRU采用验证集RankIC早停，不再固定使用最后一轮。
GRU_MAX_EPOCHS = 20
EARLY_STOP_PATIENCE = 4
RANKIC_MIN_DELTA = 1e-4

# 严格时间验证：最后20%的样本日期作为验证期，
# 验证期前留出5个交易日，防止未来5日收益标签跨越训练/验证边界。
VALIDATION_RATIO = 0.20
PURGE_TRADING_DAYS = 5
MIN_VALIDATION_DATES = 20
MIN_RANKIC_STOCKS = 30

BATCH_SIZE = 1024
BIAS = 0.008

DAILY_LOOKBACK_COUNT = SEQ_LEN + 60  # 原版实时端count=100

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

FEATURE_COLUMNS = [
    # 六个平稳化基础量价特征
    "open_gap", "intraday_ret", "high_rel", "low_rel",
    "volume_rel20", "money_rel20",
    # 收益、动量、均线、波动和技术指标
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5_rel20",
    "std_5", "std_20", "rsi_14", "macd_hist_rel", "wr_14",
    "volume_ratio",
]


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)


# =============================================================================
# 4. 20个特征：相对化和平稳化版本
# =============================================================================

def calculate_all_factors(df):
    """
    构造20维量价时序特征。

    关键调整：
    - 不再把绝对open/close/high/low直接输入模型；
    - 不再把绝对volume/money或绝对5日均量直接输入模型；
    - MACD柱除以收盘价，消除股票价格数量级差异。
    """
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)
    grouped = df.groupby("instrument", sort=False)

    # -------------------------------------------------------------------------
    # 六个平稳化基础量价特征
    # -------------------------------------------------------------------------
    df["prev_close"] = grouped["close"].shift(1)

    # 相对前收盘的开盘跳空、最高价和最低价；日内收盘收益反映K线实体。
    df["open_gap"] = df["open"] / (df["prev_close"] + 1e-8) - 1.0
    df["intraday_ret"] = df["close"] / (df["open"] + 1e-8) - 1.0
    df["high_rel"] = df["high"] / (df["prev_close"] + 1e-8) - 1.0
    df["low_rel"] = df["low"] / (df["prev_close"] + 1e-8) - 1.0

    # 成交量和成交额相对各自20日均值，消除股票规模及年代数量级差异。
    df["volume_ma20"] = grouped["volume"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["money_ma20"] = grouped["money"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["volume_rel20"] = df["volume"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["money_rel20"] = df["money"] / (df["money_ma20"] + 1e-8) - 1.0

    # -------------------------------------------------------------------------
    # 收益率与动量特征
    # -------------------------------------------------------------------------
    df["pct_chg"] = grouped["close"].pct_change()
    df["mom_5"] = grouped["close"].pct_change(5)
    df["mom_10"] = grouped["close"].pct_change(10)
    df["mom_20"] = grouped["close"].pct_change(20)

    # -------------------------------------------------------------------------
    # 均线距离特征
    # -------------------------------------------------------------------------
    df["ma5"] = grouped["close"].transform(lambda x: x.rolling(5).mean())
    df["ma10"] = grouped["close"].transform(lambda x: x.rolling(10).mean())
    df["ma20"] = grouped["close"].transform(lambda x: x.rolling(20).mean())
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    # -------------------------------------------------------------------------
    # 波动率与量价趋势特征
    # -------------------------------------------------------------------------
    df["vol_ma5"] = grouped["volume"].transform(lambda x: x.rolling(5).mean())
    df["vol_ma5_rel20"] = df["vol_ma5"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["std_5"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    # RSI 14
    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = grouped["close"].transform(lambda x: get_rsi(x, 14))

    # MACD柱状图：除以收盘价，转换成跨股票可比的相对尺度。
    df["ema12"] = grouped["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = grouped["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument", sort=False)["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist_rel"] = ((df["dif"] - df["dea"]) * 2) / (
        df["close"].abs() + 1e-8
    )

    # WR 14
    df["high_14"] = grouped["high"].transform(lambda x: x.rolling(14).max())
    df["low_14"] = grouped["low"].transform(lambda x: x.rolling(14).min())
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    # 对除零或异常行情产生的无穷值统一转为缺失值，由后续dropna剔除。
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    feat_cols = FEATURE_COLUMNS.copy()
    df.drop(
        columns=[
            "prev_close", "volume_ma20", "money_ma20",
            "ma5", "ma10", "ma20", "vol_ma5",
            "ema12", "ema26", "dif", "dea", "high_14", "low_14",
        ],
        errors="ignore",
        inplace=True,
    )
    return df, feat_cols


# =============================================================================
# 5. 模型训练：未来5日累计收益 + 原始/生成样本联合训练
# =============================================================================


def _calculate_daily_rank_ic(
    predictions,
    targets,
    sample_dates,
    sample_instruments,
):
    """
    按样本日期计算验证集横截面RankIC。

    RankIC使用预测值排名与未来5日累计收益排名之间的Pearson相关系数，
    等价于Spearman秩相关。仅保留股票数不少于MIN_RANKIC_STOCKS，
    且预测值和标签均存在横截面差异的日期。
    """
    validation_df = pd.DataFrame(
        {
            "date": pd.to_datetime(sample_dates).normalize(),
            "instrument": np.asarray(sample_instruments, dtype=str),
            "prediction": np.asarray(predictions, dtype=np.float64),
            "target": np.asarray(targets, dtype=np.float64),
        }
    )
    validation_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    validation_df.dropna(
        subset=["date", "instrument", "prediction", "target"],
        inplace=True,
    )

    daily_rank_ic = []
    for date, section in validation_df.groupby("date", sort=True):
        if len(section) < MIN_RANKIC_STOCKS:
            continue
        if (
            section["prediction"].nunique(dropna=True) < 2
            or section["target"].nunique(dropna=True) < 2
        ):
            continue

        prediction_rank = section["prediction"].rank(
            method="average",
            pct=True,
        )
        target_rank = section["target"].rank(
            method="average",
            pct=True,
        )
        rank_ic = prediction_rank.corr(target_rank)
        if pd.notna(rank_ic) and np.isfinite(rank_ic):
            daily_rank_ic.append((date, float(rank_ic), len(section)))

    if not daily_rank_ic:
        return {
            "mean": np.nan,
            "std": np.nan,
            "ir": np.nan,
            "positive_rate": np.nan,
            "days": 0,
        }

    rank_ic_values = np.asarray(
        [item[1] for item in daily_rank_ic],
        dtype=np.float64,
    )
    mean_rank_ic = float(rank_ic_values.mean())
    std_rank_ic = (
        float(rank_ic_values.std(ddof=1))
        if len(rank_ic_values) > 1
        else np.nan
    )
    rank_ic_ir = (
        mean_rank_ic / std_rank_ic
        if np.isfinite(std_rank_ic) and std_rank_ic > 1e-12
        else np.nan
    )

    return {
        "mean": mean_rank_ic,
        "std": std_rank_ic,
        "ir": rank_ic_ir,
        "positive_rate": float((rank_ic_values > 0).mean()),
        "days": int(len(rank_ic_values)),
    }


def _evaluate_gru_on_validation(
    model,
    validation_loader,
    validation_dates,
    validation_instruments,
    criterion,
):
    """在原始标准化验证样本上计算MSE和逐日横截面RankIC。"""
    model.eval()
    predictions = []
    targets = []
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for batch_x, batch_y in validation_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)

            batch_prediction = model(batch_x)
            batch_loss = criterion(batch_prediction, batch_y)

            batch_size = int(batch_x.size(0))
            total_loss += float(batch_loss.item()) * batch_size
            total_count += batch_size

            predictions.append(
                batch_prediction.detach().cpu().numpy().reshape(-1)
            )
            targets.append(batch_y.detach().cpu().numpy().reshape(-1))

    if not predictions:
        raise RuntimeError("验证集没有可用于评估的样本")

    predictions = np.concatenate(predictions)
    targets = np.concatenate(targets)

    rank_ic_metrics = _calculate_daily_rank_ic(
        predictions,
        targets,
        validation_dates,
        validation_instruments,
    )
    validation_mse = total_loss / max(total_count, 1)
    return validation_mse, rank_ic_metrics


def train_and_save_models(context, train_stocks):
    start_fetch = (
        pd.to_datetime(TRAIN_START) - pd.Timedelta(days=40)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"开始训练：股票数={len(train_stocks)}，"
        f"行情区间={start_fetch}~{TRAIN_END}"
    )

    price_df = _query_daily_bars(
        train_stocks,
        start_fetch,
        TRAIN_END,
    )
    if price_df.empty:
        raise RuntimeError("训练行情为空")

    # 对应原版 get_price(..., fq='pre')。
    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df, feat_cols = calculate_all_factors(price_df)

    context.feat_dim = len(feat_cols)

    # 标签：以当前行收盘价为起点，到未来第5个交易日收盘价的累计收益率。
    # target[t] = close[t+5] / close[t] - 1
    grouped_close = price_df.groupby("instrument", sort=False)["close"]
    price_df["target"] = grouped_close.shift(-5) / price_df["close"] - 1.0

    price_df = price_df[
        (price_df["date"] >= pd.Timestamp(TRAIN_START))
        & (price_df["date"] <= pd.Timestamp(TRAIN_END))
    ]
    price_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    price_df.dropna(subset=feat_cols + ["target"], inplace=True)

    x_list = []
    y_list = []
    sample_dates = []
    sample_instruments = []

    for instrument, group in price_df.groupby("instrument", sort=False):
        group = group.sort_values("date").reset_index(drop=True)
        if len(group) < SEQ_LEN:
            continue

        feats = group[feat_cols].to_numpy(dtype=np.float32)
        targets = group["target"].to_numpy(dtype=np.float32)
        dates = group["date"].to_numpy(dtype="datetime64[ns]")

        # 窗口最后一日end_idx直接匹配同一日的未来5日累计收益标签。
        for end_idx in range(SEQ_LEN - 1, len(group)):
            start_idx = end_idx - SEQ_LEN + 1
            x_list.append(feats[start_idx: end_idx + 1])
            y_list.append(targets[end_idx])
            sample_dates.append(dates[end_idx])
            sample_instruments.append(str(instrument))

    if not x_list:
        raise RuntimeError("未构造出有效训练样本")

    x_array = np.asarray(x_list, dtype=np.float32)
    y_array = np.asarray(y_list, dtype=np.float32).reshape(-1, 1)
    sample_dates_array = np.asarray(
        sample_dates,
        dtype="datetime64[ns]",
    )
    sample_instruments_array = np.asarray(
        sample_instruments,
        dtype=str,
    )

    del x_list, y_list, sample_dates, sample_instruments
    gc.collect()

    # -------------------------------------------------------------------------
    # 严格时间划分：按样本窗口结束日期，而不是随机划分。
    # 最后VALIDATION_RATIO的日期作为验证期；验证期之前剔除
    # PURGE_TRADING_DAYS个样本日期，避免5日标签跨越边界。
    # -------------------------------------------------------------------------
    unique_sample_dates = np.sort(np.unique(sample_dates_array))
    if len(unique_sample_dates) < MIN_VALIDATION_DATES * 2:
        raise RuntimeError(
            f"有效样本日期仅{len(unique_sample_dates)}个，"
            "不足以构造严格时间训练集和验证集"
        )

    validation_date_count = max(
        MIN_VALIDATION_DATES,
        int(np.ceil(len(unique_sample_dates) * VALIDATION_RATIO)),
    )
    validation_date_count = min(
        validation_date_count,
        len(unique_sample_dates) - MIN_VALIDATION_DATES,
    )
    validation_start_index = len(unique_sample_dates) - validation_date_count
    train_end_index = validation_start_index - PURGE_TRADING_DAYS - 1

    if train_end_index < MIN_VALIDATION_DATES - 1:
        raise RuntimeError(
            "隔离期后训练日期不足，请缩短验证比例或隔离期"
        )

    train_end_date = unique_sample_dates[train_end_index]
    validation_start_date = unique_sample_dates[validation_start_index]

    train_mask = sample_dates_array <= train_end_date
    validation_mask = sample_dates_array >= validation_start_date

    if not train_mask.any() or not validation_mask.any():
        raise RuntimeError("严格时间划分后训练集或验证集为空")

    x_train = torch.from_numpy(x_array[train_mask].copy())
    y_train = torch.from_numpy(y_array[train_mask].copy())
    x_validation = torch.from_numpy(x_array[validation_mask].copy())
    y_validation = torch.from_numpy(y_array[validation_mask].copy())

    # 保留训练样本日期，用于后续把GRU训练批次组织成完整日度横截面。
    train_dates = sample_dates_array[train_mask].copy()
    validation_dates = sample_dates_array[validation_mask].copy()
    validation_instruments = sample_instruments_array[validation_mask].copy()

    purged_dates = unique_sample_dates[
        train_end_index + 1:validation_start_index
    ]

    context.logger.info(
        "严格时间划分完成："
        f"训练期截至={pd.Timestamp(train_end_date).date()}，"
        f"验证期开始={pd.Timestamp(validation_start_date).date()}，"
        f"隔离交易日={len(purged_dates)}，"
        f"训练样本={len(x_train):,}，"
        f"验证样本={len(x_validation):,}，"
        f"验证日期={len(np.unique(validation_dates))}"
    )

    del (
        x_array,
        y_array,
        sample_dates_array,
        sample_instruments_array,
        unique_sample_dates,
        train_mask,
        validation_mask,
        purged_dates,
    )
    gc.collect()

    # 标准化参数只使用训练集，验证集不参与任何统计量估计。
    x_flat = x_train.view(-1, context.feat_dim)
    context.train_mean = x_flat.mean(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1)
    context.train_std = x_flat.std(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1) + 1e-8

    x_train = (x_train - context.train_mean) / context.train_std
    x_validation = (
        x_validation - context.train_mean
    ) / context.train_std

    gan_train_loader = DataLoader(
        TensorDataset(x_train, y_train),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )
    validation_loader = DataLoader(
        TensorDataset(x_validation, y_validation),
        batch_size=BATCH_SIZE,
        shuffle=False,
    )

    # GAN仅使用训练集，验证集不参与生成器或判别器训练。
    generator = Generator(context.feat_dim, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        context.feat_dim,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion_gan = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()
        epoch_loss_g = 0.0
        epoch_loss_d = 0.0
        epoch_samples = 0

        for real_data, _ in gan_train_loader:
            real_data = real_data.to(DEVICE)
            batch_size = int(real_data.size(0))
            real_label = torch.ones(batch_size, device=DEVICE)
            fake_label = torch.zeros(batch_size, device=DEVICE)

            discriminator.zero_grad()
            loss_d = (
                criterion_gan(discriminator(real_data), real_label)
                + criterion_gan(
                    discriminator(generator(real_data).detach()),
                    fake_label,
                )
            )
            loss_d.backward()
            opt_d.step()

            generator.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion_gan(discriminator(fake_data), real_label)
                + LAMBDA_REC
                * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

            epoch_loss_d += float(loss_d.item()) * batch_size
            epoch_loss_g += float(loss_g.item()) * batch_size
            epoch_samples += batch_size

        context.logger.info(
            f"GAN训练完成：epoch={epoch + 1}/{GAN_EPOCHS}，"
            f"loss_D={epoch_loss_d / max(epoch_samples, 1):.6f}，"
            f"loss_G={epoch_loss_g / max(epoch_samples, 1):.6f}"
        )

    # 只为训练集生成增强样本；验证集始终使用真实原始标准化样本。
    generator.eval()
    with torch.no_grad():
        x_enhanced = torch.cat(
            [
                generator(
                    x_train[i: i + BATCH_SIZE].to(DEVICE)
                ).cpu()
                for i in range(0, x_train.size(0), BATCH_SIZE)
            ],
            dim=0,
        )

    # -------------------------------------------------------------------------
    # 按日期组织GRU训练截面。
    # 原版本会把不同日期、不同股票随机混合成固定大小batch；本版本为每个
    # 交易日建立一个索引组，每次参数更新都使用该日全部有效股票。
    # 原始样本和GAN增强样本仍共同训练，并在同一日期内拼接。
    # -------------------------------------------------------------------------
    train_date_groups = {}
    for sample_index, sample_date in enumerate(train_dates):
        date_key = np.datetime64(sample_date, "ns")
        train_date_groups.setdefault(date_key, []).append(sample_index)

    train_date_keys = sorted(train_date_groups.keys())
    train_date_index_tensors = {
        date_key: torch.tensor(indices, dtype=torch.long)
        for date_key, indices in train_date_groups.items()
    }

    cross_section_sizes = np.asarray(
        [len(train_date_groups[date_key]) for date_key in train_date_keys],
        dtype=np.int64,
    )
    if len(train_date_keys) == 0:
        raise RuntimeError("未构造出可用于按日期训练的股票截面")

    context.logger.info(
        "GRU按日期训练截面已构造："
        f"交易日={len(train_date_keys)}，"
        f"平均股票数={cross_section_sizes.mean():.1f}，"
        f"最少股票数={cross_section_sizes.min()}，"
        f"最多股票数={cross_section_sizes.max()}"
    )

    del x_flat, price_df
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # -------------------------------------------------------------------------
    # GRU训练：训练批次按日期组织。每次更新使用同一天全部有效股票的
    # 原始样本与GAN增强样本；损失函数仍然只有逐样本MSE，不加入排序损失。
    # 每轮后在严格时间验证集上计算逐日横截面RankIC，并按RankIC早停。
    # -------------------------------------------------------------------------
    context.gru_model = StockGRU(
        context.feat_dim,
        64,
        2,
    ).to(DEVICE)
    optimizer_gru = optim.Adam(
        context.gru_model.parameters(),
        lr=0.001,
    )
    criterion_gru = nn.MSELoss()

    best_rank_ic = -np.inf
    best_model_state = None
    best_epoch = 0
    no_improvement_epochs = 0

    for epoch in range(GRU_MAX_EPOCHS):
        context.gru_model.train()
        total_train_loss = 0.0
        total_train_count = 0

        # 每个epoch只打乱日期顺序，不打乱同一日期内的股票。
        date_order = np.random.permutation(len(train_date_keys))

        for date_position in date_order:
            date_key = train_date_keys[int(date_position)]
            date_indices = train_date_index_tensors[date_key]

            original_x = x_train.index_select(0, date_indices)
            enhanced_x = x_enhanced.index_select(0, date_indices)
            date_y = y_train.index_select(0, date_indices)

            # 同一日期的原始样本与增强样本合并；每只股票标签复制一次。
            batch_x = torch.cat([original_x, enhanced_x], dim=0).to(DEVICE)
            batch_y = torch.cat([date_y, date_y], dim=0).to(DEVICE)

            optimizer_gru.zero_grad()
            prediction = context.gru_model(batch_x)
            loss = criterion_gru(prediction, batch_y)
            loss.backward()
            optimizer_gru.step()

            batch_size = int(batch_x.size(0))
            total_train_loss += float(loss.item()) * batch_size
            total_train_count += batch_size

        train_mse = total_train_loss / max(total_train_count, 1)
        validation_mse, rank_ic_metrics = _evaluate_gru_on_validation(
            context.gru_model,
            validation_loader,
            validation_dates,
            validation_instruments,
            criterion_gru,
        )
        mean_rank_ic = rank_ic_metrics["mean"]

        improved = (
            np.isfinite(mean_rank_ic)
            and mean_rank_ic > best_rank_ic + RANKIC_MIN_DELTA
        )

        if improved:
            best_rank_ic = float(mean_rank_ic)
            best_epoch = epoch + 1
            no_improvement_epochs = 0
            best_model_state = {
                key: value.detach().cpu().clone()
                for key, value
                in context.gru_model.state_dict().items()
            }
        else:
            no_improvement_epochs += 1

        context.logger.info(
            f"GRU epoch={epoch + 1}/{GRU_MAX_EPOCHS}，"
            f"train_MSE={train_mse:.8f}，"
            f"val_MSE={validation_mse:.8f}，"
            f"val_RankIC={mean_rank_ic:.6f}，"
            f"RankICIR={rank_ic_metrics['ir']:.4f}，"
            f"RankIC胜率={rank_ic_metrics['positive_rate']:.4f}，"
            f"有效截面={rank_ic_metrics['days']}，"
            f"最佳epoch={best_epoch}，"
            f"未改善轮数={no_improvement_epochs}"
        )

        if no_improvement_epochs >= EARLY_STOP_PATIENCE:
            context.logger.info(
                f"触发RankIC早停：连续{EARLY_STOP_PATIENCE}轮未改善"
            )
            break

    if best_model_state is None:
        raise RuntimeError(
            "验证集未计算出有效RankIC，无法选择最佳GRU模型"
        )

    context.gru_model.load_state_dict(best_model_state)
    context.gru_model.to(DEVICE)
    context.gru_model.eval()
    context.model_ready = True
    context.best_gru_epoch = best_epoch
    context.best_validation_rank_ic = best_rank_ic
    context.validation_start_date = pd.Timestamp(
        validation_start_date
    ).strftime("%Y-%m-%d")

    del (
        x_train,
        x_validation,
        x_enhanced,
        y_train,
        y_validation,
        train_dates,
        train_date_groups,
        train_date_keys,
        train_date_index_tensors,
        cross_section_sizes,
        generator,
        discriminator,
        best_model_state,
        validation_dates,
        validation_instruments,
    )
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    context.logger.info(
        "GAN + GRU模型已就绪："
        "目标=未来5日累计收益，"
        "GRU训练=按日期完整截面（原始样本+生成样本），"
        f"最佳epoch={best_epoch}，"
        f"最佳验证RankIC={best_rank_ic:.6f}"
    )


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None

    # 权重数据在bigtrader.run之前只读取一次，同时用于限定回测股票池。
    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.logger.info(
        f"已加载沪深300权重缓存：{len(context.index_weights_df):,}行，"
        f"股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    # 训练股票池：训练结束日的沪深300成分股。
    train_weights = _weights_asof(context.index_weights_df, TRAIN_END)
    train_stocks = train_weights["instrument"].tolist()
    if not train_stocks:
        raise RuntimeError(f"无法取得{TRAIN_END}沪深300成分股")

    train_and_save_models(context, train_stocks)

    # 一次性读取回测期可能用到的全部日线原始数据；每周预测时仍按原版逻辑
    # 截取当期股票最近100个完整日线，并重新做末端前复权与因子计算。
    backtest_sections = context.index_weights_df[
        (context.index_weights_df["date"] >= pd.Timestamp(context.start_date) - pd.Timedelta(days=180))
        & (context.index_weights_df["date"] <= pd.Timestamp(context.end_date))
    ]
    universe = sorted(
        backtest_sections["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300股票并集为空")

    daily_start = (
        pd.Timestamp(context.start_date) - pd.Timedelta(days=220)
    ).strftime("%Y-%m-%d")

    context.logger.info(
        f"读取回测期日线缓存：{daily_start}~{context.end_date}，"
        f"股票并集={len(universe)}"
    )
    daily_df = _query_daily_bars(
        universe,
        daily_start,
        context.end_date,
    )
    if daily_df.empty:
        raise RuntimeError("回测期日线缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }

    del daily_df, backtest_sections
    gc.collect()


def _build_realtime_factor_data(context, stocks, current_date):
    """
    复刻原版：
        get_price(stocks, end_date=current_date,
                  count=SEQ_LEN+60, frequency='daily', fq='pre')

    日频回测在周五调仓，只使用current_date之前已经完整结束的日线。
    因此信号在周五交易前已经可得，不使用周五收盘后信息。
    """
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue

        # 原版count=100；从完整日线中取最近100条。
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if section.empty:
            continue
        pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df, feat_cols = calculate_all_factors(realtime_df)
    realtime_df.dropna(inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(
        context,
        stocks,
        current_date,
    )
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    # 批量推理只减少Python循环和模型调用次数，不改变每只股票的输入、
    # 标准化方式、模型结构或预测结果含义。预测值对应未来5日累计收益，实时推理仍不经过Generator。
    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )

    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 回测静态数据准备
# =============================================================================

def _prepare_backtest_static_data():
    """
    在启动BigTrader前读取一次指数权重，并据此限定回测股票池。

    这样日频引擎只加载回测期沪深300历史成分股并集，避免在未指定
    instruments时准备全市场行情。
    """
    weight_start = (
        min(pd.Timestamp(TRAIN_END), pd.Timestamp(BACKTEST_START))
        - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(weight_start, BACKTEST_END)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        using_official = False

    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    return weights, using_official, universe


(
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
) = _prepare_backtest_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

```

该版模型效果与改进版2效果类似，但从逻辑上看，这版的逻辑则更为合理，有理由相信在长期的市场环境下该版模型的效果会更好，同时，目前的模型固定了一个训练时间，模型不会滚动向前训练来预测新的数据，会存在滞后性，下一版模型将重点改进这个部分

## 改进版模型5

本版本将原来的单次固定训练改为滚动训练。

只需设置回测区间、训练窗口长度和模型有效期，代码会自动生成多个训练窗口，并为每个时间段训练对应模型。

例如，设置16个月训练窗口、4个月模型有效期后，每个模型会使用生效日前16个月的数据训练，并负责未来4个月的预测。有效期结束后，训练窗口自动向前滚动并切换到新模型。

每个滚动窗口都会独立完成特征标准化、GAN训练、GRU训练、时间验证集划分和RankIC早停。回测期间每周调仓时，代码会自动选择当前日期对应的有效模型。

该改进能够减少模型长期固定带来的失效问题，使模型持续适应最新的市场状态，同时严格避免使用未来数据。

```python
# -*- coding: utf-8 -*-
"""
GAN + GRU 沪深300指数增强策略
BigQuant DAI + BigTrader 原生日频滚动训练回测版

本版以“回测区间、训练窗口月数、模型有效月数”为核心参数，自动生成
Walk-Forward滚动模型计划：
1. 每个模型使用生效日前固定长度的历史窗口训练；
2. 标签为窗口末日到未来第5个交易日的累计收益率；
3. 严格按时间划分训练集/验证集，并保留5个交易日隔离期；
4. GAN仅使用训练集，GRU使用原始样本与生成样本共同训练；
5. GRU按日期完整横截面组织训练，损失仍为MSE；
6. 每轮在验证集计算逐日横截面RankIC，并按RankIC早停；
7. 所有滚动模型在BigTrader初始化阶段依次训练，回测时按日期自动切换；
8. 每个模型有效期内仍然每周五使用最新40日数据重新预测和调仓。

用户只需设置：
- BACKTEST_START / BACKTEST_END
- TRAIN_WINDOW_MONTHS
- MODEL_VALID_MONTHS

BigQuant接口：
- from bigquant import bigtrader, dai
- bigtrader.run(..., frequency=bigtrader.Frequency.DAILY)
- cn_stock_bar1d日线数据，amount映射为成交额money
- cn_stock_index_weight指数权重
- context.order_target_percent目标仓位下单
"""

from datetime import datetime
import gc
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from bigquant import bigtrader, dai

warnings.filterwarnings("ignore")

# 当前BigQuant DAI环境不提供dai.get_datasource_schema，
# 因此本策略只使用官方文档明确给出的固定表名和字段，不做运行时schema探测。


# =============================================================================
# 1. 参数：滚动训练计划由核心时间参数自动生成
# =============================================================================

INDEX_CODE = "000300.SH"

# 回测区间。
BACKTEST_START = "2024-05-02"
BACKTEST_END = "2026-07-13"
CAPITAL_BASE = 1_000_000

# 自动滚动参数：参考此前“约16个月训练、约4个月样本外使用”的较好结果。
TRAIN_WINDOW_MONTHS = 16
MODEL_VALID_MONTHS = 4

# 为训练窗口开始日前补充历史数据，用于40日序列和滚动指标预热。
TRAIN_FETCH_BUFFER_DAYS = 120

SEQ_LEN = 40
LATENT_DIM = 10
D_HIDDEN = 32
LAMBDA_REC = 0.5

GAN_EPOCHS = 4

# GRU采用验证集RankIC早停。
GRU_MAX_EPOCHS = 20
EARLY_STOP_PATIENCE = 4
RANKIC_MIN_DELTA = 1e-4

# 严格时间验证：最后20%的样本日期作为验证期；验证期前留出5个交易日。
VALIDATION_RATIO = 0.20
PURGE_TRADING_DAYS = 5
MIN_VALIDATION_DATES = 20
MIN_RANKIC_STOCKS = 30

BATCH_SIZE = 1024
BIAS = 0.008

DAILY_LOOKBACK_COUNT = SEQ_LEN + 60
BASE_RANDOM_SEED = 42

try:
    torch.set_num_threads(4)
except Exception:
    pass

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

FEATURE_COLUMNS = [
    "open_gap", "intraday_ret", "high_rel", "low_rel",
    "volume_rel20", "money_rel20",
    "pct_chg", "mom_5", "mom_10", "mom_20",
    "dist_ma5", "dist_ma10", "dist_ma20", "vol_ma5_rel20",
    "std_5", "std_20", "rsi_14", "macd_hist_rel", "wr_14",
    "volume_ratio",
]


def _build_rolling_model_plan():
    """
    根据回测区间、训练窗口月数和模型有效月数自动生成滚动计划。

    有效期按自然月分段。例如回测从2024-05-02开始、有效期4个月：
        模型1有效期：2024-05-02 ~ 2024-08-31
        模型2有效期：2024-09-01 ~ 2024-12-31

    训练窗口同样按月对齐。例如模型1使用16个月：
        训练窗口：2023-01-01 ~ 2024-05-01
    最后5个交易日因future_return_5不可知，会在构造标签时自然被剔除。
    """
    backtest_start = pd.Timestamp(BACKTEST_START).normalize()
    backtest_end = pd.Timestamp(BACKTEST_END).normalize()

    if backtest_start > backtest_end:
        raise ValueError("BACKTEST_START不能晚于BACKTEST_END")
    if TRAIN_WINDOW_MONTHS <= 0 or MODEL_VALID_MONTHS <= 0:
        raise ValueError("TRAIN_WINDOW_MONTHS和MODEL_VALID_MONTHS必须为正整数")

    plan = []
    effective_start = backtest_start
    model_id = 1

    while effective_start <= backtest_end:
        effective_start_period = effective_start.to_period("M")
        effective_end = (
            effective_start_period + (MODEL_VALID_MONTHS - 1)
        ).end_time.normalize()
        effective_end = min(effective_end, backtest_end)

        train_start = (
            effective_start_period - TRAIN_WINDOW_MONTHS
        ).start_time.normalize()
        training_data_end = effective_start - pd.Timedelta(days=1)

        plan.append(
            {
                "model_id": model_id,
                "train_start": train_start,
                "training_data_end": training_data_end,
                "effective_start": effective_start,
                "effective_end": effective_end,
            }
        )

        effective_start = effective_end + pd.Timedelta(days=1)
        model_id += 1

    return plan


# =============================================================================
# 2. 模型结构：逐项复刻原版
# =============================================================================

class Generator(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.fc_noise = nn.Linear(latent_dim, input_dim)
        self.rnn = nn.GRU(input_dim, input_dim, num_layers=1, batch_first=True)

    def forward(self, x):
        bs, seq_len, _ = x.size()
        noise = torch.randn(bs, seq_len, self.latent_dim, device=x.device)
        return self.rnn(x + self.fc_noise(noise))[0]


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_dim, hidden_dim):
        super().__init__()
        self.rnn = nn.GRU(input_dim, hidden_dim, num_layers=1, batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim * seq_len, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        bs, _, _ = x.size()
        out, _ = self.rnn(x)
        return self.fc(out.contiguous().view(bs, -1)).squeeze(-1)


class StockGRU(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2):
        super().__init__()
        self.gru = nn.GRU(
            input_dim,
            hidden_dim,
            num_layers,
            batch_first=True,
            dropout=0.2,
        )
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        return self.fc(self.gru(x)[1][-1])


# =============================================================================
# 3. BigQuant数据辅助函数
# =============================================================================

def _query_daily_bars(stocks, start_date, end_date):
    """
    读取BigQuant后复权日线。

    BigQuant官方cn_stock_bar1d字段中直接包含adjust_factor，原版money映射为amount。
    不再调用当前环境不存在的dai.get_datasource_schema。
    """
    if not stocks:
        return pd.DataFrame()

    sql = """
    SELECT
        date,
        instrument,
        open,
        close,
        high,
        low,
        volume,
        amount AS money,
        adjust_factor
    FROM cn_stock_bar1d
    WHERE instrument IN $stocks
    ORDER BY instrument, date
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"stocks": list(stocks)},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)

    numeric_cols = [
        "open", "close", "high", "low", "volume", "money",
        "adjust_factor",
    ]
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return (
        df.sort_values(["instrument", "date"])
        .drop_duplicates(["instrument", "date"], keep="last")
        .reset_index(drop=True)
    )


def _convert_to_endpoint_pre_adjusted(df):
    """
    将BigQuant后复权OHLC转换为以每只股票当前数据末日为基准的前复权口径。

    若后复权价 = 真实价 × 当日累计复权因子，则：
        末日基准前复权价 = 后复权价 / 末日累计复权因子

    对同一股票整个窗口除以同一个末日因子，不改变窗口内收益率、均线距离、
    RSI、MACD、WR等比例型特征，并使绝对OHLC更贴近原策略fq='pre'口径。
    """
    if df.empty:
        return df

    result = df.copy()
    result = result.sort_values(["instrument", "date"]).reset_index(drop=True)

    def _adjust_one(group):
        group = group.copy()
        factors = pd.to_numeric(
            group["adjust_factor"], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)
        factors = factors[factors.notna() & (factors != 0)]
        endpoint_factor = float(factors.iloc[-1]) if not factors.empty else 1.0
        for col in ["open", "close", "high", "low"]:
            group[col] = group[col] / endpoint_factor
        return group

    result = (
        result.groupby("instrument", group_keys=False, sort=False)
        .apply(_adjust_one)
        .reset_index(drop=True)
    )
    result.drop(columns=["adjust_factor"], errors="ignore", inplace=True)
    return result


def _query_official_index_weights(start_date, end_date):
    """
    读取沪深300官方指数权重。

    BigQuant官方表结构：
        表名：cn_stock_index_weight
        instrument：指数代码
        member_code：成分股代码
        weight：成分股权重
    """
    sql = """
    SELECT
        date,
        member_code AS instrument,
        weight
    FROM cn_stock_index_weight
    WHERE instrument = $index_code
    ORDER BY date, member_code
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
        params={"index_code": INDEX_CODE},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df["weight"] = pd.to_numeric(df["weight"], errors="coerce")
    return (
        df.dropna(subset=["date", "instrument", "weight"])
        .sort_values(["date", "instrument"])
        .drop_duplicates(["date", "instrument"], keep="last")
        .reset_index(drop=True)
    )

def _query_equal_weight_hs300(start_date, end_date):
    """官方权重失败时，按原版逻辑退化为每日沪深300成分股等权。"""
    sql = """
    SELECT date, instrument
    FROM cn_stock_prefactors
    WHERE is_hs300 = 1
    ORDER BY date, instrument
    """

    df = dai.query(
        sql,
        filters={"date": [start_date, end_date]},
    ).df()

    if df.empty:
        return df

    df["date"] = pd.to_datetime(df["date"]).dt.normalize()
    df["instrument"] = df["instrument"].astype(str)
    df = df.drop_duplicates(["date", "instrument"], keep="last")
    count = df.groupby("date")["instrument"].transform("count")
    df["weight"] = 1.0 / count
    return df[["date", "instrument", "weight"]].reset_index(drop=True)


def _load_index_weights_with_fallback(start_date, end_date, logger):
    try:
        weights = _query_official_index_weights(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        logger.info(
            f"已读取沪深300官方权重：{len(weights):,}行，"
            f"{weights['date'].nunique()}个权重截面"
        )
        return weights, True
    except Exception as exc:
        logger.warning(
            "官方指数权重读取失败，按原版兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(start_date, end_date)
        if weights.empty:
            raise RuntimeError("官方权重与等权回退均未取得数据") from exc
        return weights, False


def _weights_asof(weights_df, target_date):
    """取得不晚于目标日的最新指数权重截面。"""
    if weights_df is None or weights_df.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    target_ts = pd.Timestamp(target_date).normalize()
    available_dates = weights_df.loc[
        weights_df["date"] <= target_ts,
        "date",
    ]
    if available_dates.empty:
        return pd.DataFrame(columns=["instrument", "weight"])

    latest_date = available_dates.max()
    section = weights_df.loc[
        weights_df["date"] == latest_date,
        ["instrument", "weight"],
    ].copy()

    section = section.drop_duplicates("instrument", keep="last")
    section["weight"] = pd.to_numeric(section["weight"], errors="coerce")
    section = section.dropna(subset=["weight"])

    if section.empty:
        return section

    # 兼容官方表以100或1为权重总和的两种口径。
    if float(section["weight"].sum()) > 2.0:
        section["weight"] = section["weight"] / 100.0

    return section.reset_index(drop=True)


# =============================================================================
# 4. 20个特征：相对化和平稳化版本
# =============================================================================

def calculate_all_factors(df):
    """
    构造20维量价时序特征。

    关键调整：
    - 不再把绝对open/close/high/low直接输入模型；
    - 不再把绝对volume/money或绝对5日均量直接输入模型；
    - MACD柱除以收盘价，消除股票价格数量级差异。
    """
    df = df.sort_values(["instrument", "date"]).reset_index(drop=True)
    grouped = df.groupby("instrument", sort=False)

    # -------------------------------------------------------------------------
    # 六个平稳化基础量价特征
    # -------------------------------------------------------------------------
    df["prev_close"] = grouped["close"].shift(1)

    # 相对前收盘的开盘跳空、最高价和最低价；日内收盘收益反映K线实体。
    df["open_gap"] = df["open"] / (df["prev_close"] + 1e-8) - 1.0
    df["intraday_ret"] = df["close"] / (df["open"] + 1e-8) - 1.0
    df["high_rel"] = df["high"] / (df["prev_close"] + 1e-8) - 1.0
    df["low_rel"] = df["low"] / (df["prev_close"] + 1e-8) - 1.0

    # 成交量和成交额相对各自20日均值，消除股票规模及年代数量级差异。
    df["volume_ma20"] = grouped["volume"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["money_ma20"] = grouped["money"].transform(
        lambda x: x.rolling(20).mean()
    )
    df["volume_rel20"] = df["volume"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["money_rel20"] = df["money"] / (df["money_ma20"] + 1e-8) - 1.0

    # -------------------------------------------------------------------------
    # 收益率与动量特征
    # -------------------------------------------------------------------------
    df["pct_chg"] = grouped["close"].pct_change()
    df["mom_5"] = grouped["close"].pct_change(5)
    df["mom_10"] = grouped["close"].pct_change(10)
    df["mom_20"] = grouped["close"].pct_change(20)

    # -------------------------------------------------------------------------
    # 均线距离特征
    # -------------------------------------------------------------------------
    df["ma5"] = grouped["close"].transform(lambda x: x.rolling(5).mean())
    df["ma10"] = grouped["close"].transform(lambda x: x.rolling(10).mean())
    df["ma20"] = grouped["close"].transform(lambda x: x.rolling(20).mean())
    df["dist_ma5"] = (df["close"] - df["ma5"]) / (df["ma5"] + 1e-8)
    df["dist_ma10"] = (df["close"] - df["ma10"]) / (df["ma10"] + 1e-8)
    df["dist_ma20"] = (df["close"] - df["ma20"]) / (df["ma20"] + 1e-8)

    # -------------------------------------------------------------------------
    # 波动率与量价趋势特征
    # -------------------------------------------------------------------------
    df["vol_ma5"] = grouped["volume"].transform(lambda x: x.rolling(5).mean())
    df["vol_ma5_rel20"] = df["vol_ma5"] / (df["volume_ma20"] + 1e-8) - 1.0
    df["std_5"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(5).std()
    )
    df["std_20"] = df.groupby("instrument", sort=False)["pct_chg"].transform(
        lambda x: x.rolling(20).std()
    )
    df["volume_ratio"] = df["volume"] / (df["vol_ma5"] + 1e-8)

    # RSI 14
    def get_rsi(series, period=14):
        delta = series.diff()
        gain = delta.where(delta > 0, 0).rolling(period).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(period).mean()
        return 100 - (100 / (1 + gain / (loss + 1e-8) + 1e-8))

    df["rsi_14"] = grouped["close"].transform(lambda x: get_rsi(x, 14))

    # MACD柱状图：除以收盘价，转换成跨股票可比的相对尺度。
    df["ema12"] = grouped["close"].transform(
        lambda x: x.ewm(span=12, adjust=False).mean()
    )
    df["ema26"] = grouped["close"].transform(
        lambda x: x.ewm(span=26, adjust=False).mean()
    )
    df["dif"] = df["ema12"] - df["ema26"]
    df["dea"] = df.groupby("instrument", sort=False)["dif"].transform(
        lambda x: x.ewm(span=9, adjust=False).mean()
    )
    df["macd_hist_rel"] = ((df["dif"] - df["dea"]) * 2) / (
        df["close"].abs() + 1e-8
    )

    # WR 14
    df["high_14"] = grouped["high"].transform(lambda x: x.rolling(14).max())
    df["low_14"] = grouped["low"].transform(lambda x: x.rolling(14).min())
    df["wr_14"] = -100 * (df["high_14"] - df["close"]) / (
        df["high_14"] - df["low_14"] + 1e-8
    )

    # 对除零或异常行情产生的无穷值统一转为缺失值，由后续dropna剔除。
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    feat_cols = FEATURE_COLUMNS.copy()
    df.drop(
        columns=[
            "prev_close", "volume_ma20", "money_ma20",
            "ma5", "ma10", "ma20", "vol_ma5",
            "ema12", "ema26", "dif", "dea", "high_14", "low_14",
        ],
        errors="ignore",
        inplace=True,
    )
    return df, feat_cols


# =============================================================================
# 5. 模型训练：未来5日累计收益 + 原始/生成样本联合训练
# =============================================================================


def _slice_cached_daily_bars(context, stocks, start_date, end_date):
    """从一次性日线缓存中切取某个滚动模型所需的行情。"""
    start_ts = pd.Timestamp(start_date).normalize()
    end_ts = pd.Timestamp(end_date).normalize()
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(str(stock))
        if group is None or group.empty:
            continue
        section = group.loc[
            (group["date"] >= start_ts) & (group["date"] <= end_ts)
        ]
        if not section.empty:
            pieces.append(section.copy())

    if not pieces:
        return pd.DataFrame()
    return pd.concat(pieces, ignore_index=True)


def _calculate_daily_rank_ic(
    predictions,
    targets,
    sample_dates,
    sample_instruments,
):
    """
    按样本日期计算验证集横截面RankIC。

    RankIC使用预测值排名与未来5日累计收益排名之间的Pearson相关系数，
    等价于Spearman秩相关。仅保留股票数不少于MIN_RANKIC_STOCKS，
    且预测值和标签均存在横截面差异的日期。
    """
    validation_df = pd.DataFrame(
        {
            "date": pd.to_datetime(sample_dates).normalize(),
            "instrument": np.asarray(sample_instruments, dtype=str),
            "prediction": np.asarray(predictions, dtype=np.float64),
            "target": np.asarray(targets, dtype=np.float64),
        }
    )
    validation_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    validation_df.dropna(
        subset=["date", "instrument", "prediction", "target"],
        inplace=True,
    )

    daily_rank_ic = []
    for date, section in validation_df.groupby("date", sort=True):
        if len(section) < MIN_RANKIC_STOCKS:
            continue
        if (
            section["prediction"].nunique(dropna=True) < 2
            or section["target"].nunique(dropna=True) < 2
        ):
            continue

        prediction_rank = section["prediction"].rank(
            method="average",
            pct=True,
        )
        target_rank = section["target"].rank(
            method="average",
            pct=True,
        )
        rank_ic = prediction_rank.corr(target_rank)
        if pd.notna(rank_ic) and np.isfinite(rank_ic):
            daily_rank_ic.append((date, float(rank_ic), len(section)))

    if not daily_rank_ic:
        return {
            "mean": np.nan,
            "std": np.nan,
            "ir": np.nan,
            "positive_rate": np.nan,
            "days": 0,
        }

    rank_ic_values = np.asarray(
        [item[1] for item in daily_rank_ic],
        dtype=np.float64,
    )
    mean_rank_ic = float(rank_ic_values.mean())
    std_rank_ic = (
        float(rank_ic_values.std(ddof=1))
        if len(rank_ic_values) > 1
        else np.nan
    )
    rank_ic_ir = (
        mean_rank_ic / std_rank_ic
        if np.isfinite(std_rank_ic) and std_rank_ic > 1e-12
        else np.nan
    )

    return {
        "mean": mean_rank_ic,
        "std": std_rank_ic,
        "ir": rank_ic_ir,
        "positive_rate": float((rank_ic_values > 0).mean()),
        "days": int(len(rank_ic_values)),
    }


def _evaluate_gru_on_validation(
    model,
    validation_loader,
    validation_dates,
    validation_instruments,
    criterion,
):
    """在原始标准化验证样本上计算MSE和逐日横截面RankIC。"""
    model.eval()
    predictions = []
    targets = []
    total_loss = 0.0
    total_count = 0

    with torch.no_grad():
        for batch_x, batch_y in validation_loader:
            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)

            batch_prediction = model(batch_x)
            batch_loss = criterion(batch_prediction, batch_y)

            batch_size = int(batch_x.size(0))
            total_loss += float(batch_loss.item()) * batch_size
            total_count += batch_size

            predictions.append(
                batch_prediction.detach().cpu().numpy().reshape(-1)
            )
            targets.append(batch_y.detach().cpu().numpy().reshape(-1))

    if not predictions:
        raise RuntimeError("验证集没有可用于评估的样本")

    predictions = np.concatenate(predictions)
    targets = np.concatenate(targets)

    rank_ic_metrics = _calculate_daily_rank_ic(
        predictions,
        targets,
        validation_dates,
        validation_instruments,
    )
    validation_mse = total_loss / max(total_count, 1)
    return validation_mse, rank_ic_metrics


def _train_single_rolling_model(context, model_spec):
    """训练一个滚动窗口模型并返回可在对应有效期加载的模型包。"""
    model_id = int(model_spec["model_id"])
    train_start = pd.Timestamp(model_spec["train_start"]).normalize()
    training_data_end = pd.Timestamp(
        model_spec["training_data_end"]
    ).normalize()
    effective_start = pd.Timestamp(model_spec["effective_start"]).normalize()
    effective_end = pd.Timestamp(model_spec["effective_end"]).normalize()

    # 每个滚动模型使用生效日前最新的沪深300成分股作为训练股票池。
    train_weights = _weights_asof(
        context.index_weights_df,
        training_data_end,
    )
    train_stocks = train_weights["instrument"].astype(str).tolist()
    if not train_stocks:
        raise RuntimeError(
            f"模型{model_id}无法取得{training_data_end.date()}之前的沪深300成分股"
        )

    seed = BASE_RANDOM_SEED + model_id
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    start_fetch = train_start - pd.Timedelta(days=TRAIN_FETCH_BUFFER_DAYS)
    context.logger.info(
        f"[滚动模型{model_id}] 开始训练："
        f"训练窗口={train_start.date()}~{training_data_end.date()}，"
        f"有效期={effective_start.date()}~{effective_end.date()}，"
        f"股票数={len(train_stocks)}，"
        f"行情预热起点={start_fetch.date()}"
    )

    price_df = _slice_cached_daily_bars(
        context,
        train_stocks,
        start_fetch,
        training_data_end,
    )
    if price_df.empty:
        raise RuntimeError(f"滚动模型{model_id}训练行情为空")

    price_df = _convert_to_endpoint_pre_adjusted(price_df)
    price_df, feat_cols = calculate_all_factors(price_df)
    feat_dim = len(feat_cols)

    # 标签：窗口末日到未来第5个交易日收盘的累计收益率。
    grouped_close = price_df.groupby("instrument", sort=False)["close"]
    price_df["target"] = (
        grouped_close.shift(-5) / price_df["close"] - 1.0
    )

    price_df = price_df.loc[
        (price_df["date"] >= train_start)
        & (price_df["date"] <= training_data_end)
    ].copy()
    price_df.replace([np.inf, -np.inf], np.nan, inplace=True)
    price_df.dropna(subset=feat_cols + ["target"], inplace=True)

    x_list = []
    y_list = []
    sample_dates = []
    sample_instruments = []

    for instrument, group in price_df.groupby("instrument", sort=False):
        group = group.sort_values("date").reset_index(drop=True)
        if len(group) < SEQ_LEN:
            continue

        feats = group[feat_cols].to_numpy(dtype=np.float32)
        targets = group["target"].to_numpy(dtype=np.float32)
        dates = group["date"].to_numpy(dtype="datetime64[ns]")

        for end_idx in range(SEQ_LEN - 1, len(group)):
            start_idx = end_idx - SEQ_LEN + 1
            x_list.append(feats[start_idx: end_idx + 1])
            y_list.append(targets[end_idx])
            sample_dates.append(dates[end_idx])
            sample_instruments.append(str(instrument))

    if not x_list:
        raise RuntimeError(f"滚动模型{model_id}未构造出有效训练样本")

    x_array = np.asarray(x_list, dtype=np.float32)
    y_array = np.asarray(y_list, dtype=np.float32).reshape(-1, 1)
    sample_dates_array = np.asarray(sample_dates, dtype="datetime64[ns]")
    sample_instruments_array = np.asarray(sample_instruments, dtype=str)

    del x_list, y_list, sample_dates, sample_instruments
    gc.collect()

    # 严格时间划分：最后20%样本日期为验证期，前置5个交易日隔离。
    unique_sample_dates = np.sort(np.unique(sample_dates_array))
    if len(unique_sample_dates) < MIN_VALIDATION_DATES * 2:
        raise RuntimeError(
            f"滚动模型{model_id}有效样本日期仅{len(unique_sample_dates)}个，"
            "不足以构造严格时间训练集和验证集"
        )

    validation_date_count = max(
        MIN_VALIDATION_DATES,
        int(np.ceil(len(unique_sample_dates) * VALIDATION_RATIO)),
    )
    validation_date_count = min(
        validation_date_count,
        len(unique_sample_dates) - MIN_VALIDATION_DATES,
    )
    validation_start_index = len(unique_sample_dates) - validation_date_count
    train_end_index = validation_start_index - PURGE_TRADING_DAYS - 1

    if train_end_index < MIN_VALIDATION_DATES - 1:
        raise RuntimeError(
            f"滚动模型{model_id}隔离期后训练日期不足"
        )

    inner_train_end_date = unique_sample_dates[train_end_index]
    validation_start_date = unique_sample_dates[validation_start_index]

    train_mask = sample_dates_array <= inner_train_end_date
    validation_mask = sample_dates_array >= validation_start_date
    if not train_mask.any() or not validation_mask.any():
        raise RuntimeError(
            f"滚动模型{model_id}严格时间划分后训练集或验证集为空"
        )

    x_train = torch.from_numpy(x_array[train_mask].copy())
    y_train = torch.from_numpy(y_array[train_mask].copy())
    x_validation = torch.from_numpy(x_array[validation_mask].copy())
    y_validation = torch.from_numpy(y_array[validation_mask].copy())

    train_dates = sample_dates_array[train_mask].copy()
    validation_dates = sample_dates_array[validation_mask].copy()
    validation_instruments = sample_instruments_array[validation_mask].copy()

    purged_dates = unique_sample_dates[
        train_end_index + 1:validation_start_index
    ]

    context.logger.info(
        f"[滚动模型{model_id}] 时间划分："
        f"内部训练截至={pd.Timestamp(inner_train_end_date).date()}，"
        f"验证开始={pd.Timestamp(validation_start_date).date()}，"
        f"隔离交易日={len(purged_dates)}，"
        f"训练样本={len(x_train):,}，验证样本={len(x_validation):,}，"
        f"验证日期={len(np.unique(validation_dates))}"
    )

    del (
        x_array,
        y_array,
        sample_dates_array,
        sample_instruments_array,
        unique_sample_dates,
        train_mask,
        validation_mask,
        purged_dates,
    )
    gc.collect()

    # 标准化参数只使用该滚动窗口的内部训练集。
    x_flat = x_train.view(-1, feat_dim)
    train_mean = x_flat.mean(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1)
    train_std = x_flat.std(
        dim=0,
        keepdim=True,
    ).view(1, 1, -1) + 1e-8

    x_train = (x_train - train_mean) / train_std
    x_validation = (x_validation - train_mean) / train_std

    gan_train_loader = DataLoader(
        TensorDataset(x_train, y_train),
        batch_size=BATCH_SIZE,
        shuffle=True,
    )
    validation_loader = DataLoader(
        TensorDataset(x_validation, y_validation),
        batch_size=BATCH_SIZE,
        shuffle=False,
    )

    generator = Generator(feat_dim, LATENT_DIM).to(DEVICE)
    discriminator = Discriminator(
        SEQ_LEN,
        feat_dim,
        D_HIDDEN,
    ).to(DEVICE)
    opt_g = optim.Adam(generator.parameters(), lr=0.0002)
    opt_d = optim.Adam(discriminator.parameters(), lr=0.0002)
    criterion_gan = nn.BCELoss()

    for epoch in range(GAN_EPOCHS):
        generator.train()
        discriminator.train()
        epoch_loss_g = 0.0
        epoch_loss_d = 0.0
        epoch_samples = 0

        for real_data, _ in gan_train_loader:
            real_data = real_data.to(DEVICE)
            batch_size = int(real_data.size(0))
            real_label = torch.ones(batch_size, device=DEVICE)
            fake_label = torch.zeros(batch_size, device=DEVICE)

            discriminator.zero_grad()
            loss_d = (
                criterion_gan(discriminator(real_data), real_label)
                + criterion_gan(
                    discriminator(generator(real_data).detach()),
                    fake_label,
                )
            )
            loss_d.backward()
            opt_d.step()

            generator.zero_grad()
            fake_data = generator(real_data)
            loss_g = (
                criterion_gan(discriminator(fake_data), real_label)
                + LAMBDA_REC
                * nn.functional.l1_loss(fake_data, real_data)
            )
            loss_g.backward()
            opt_g.step()

            epoch_loss_d += float(loss_d.item()) * batch_size
            epoch_loss_g += float(loss_g.item()) * batch_size
            epoch_samples += batch_size

        context.logger.info(
            f"[滚动模型{model_id}] GAN epoch={epoch + 1}/{GAN_EPOCHS}，"
            f"loss_D={epoch_loss_d / max(epoch_samples, 1):.6f}，"
            f"loss_G={epoch_loss_g / max(epoch_samples, 1):.6f}"
        )

    # 仅增强训练集；验证集始终保持真实原始样本。
    generator.eval()
    with torch.no_grad():
        x_enhanced = torch.cat(
            [
                generator(
                    x_train[i:i + BATCH_SIZE].to(DEVICE)
                ).cpu()
                for i in range(0, x_train.size(0), BATCH_SIZE)
            ],
            dim=0,
        )

    # 按样本日期组织GRU训练截面。
    train_date_groups = {}
    for sample_index, sample_date in enumerate(train_dates):
        date_key = np.datetime64(sample_date, "ns")
        train_date_groups.setdefault(date_key, []).append(sample_index)

    train_date_keys = sorted(train_date_groups.keys())
    train_date_index_tensors = {
        date_key: torch.tensor(indices, dtype=torch.long)
        for date_key, indices in train_date_groups.items()
    }
    cross_section_sizes = np.asarray(
        [len(train_date_groups[key]) for key in train_date_keys],
        dtype=np.int64,
    )
    if len(train_date_keys) == 0:
        raise RuntimeError(
            f"滚动模型{model_id}未构造出按日期训练截面"
        )

    context.logger.info(
        f"[滚动模型{model_id}] GRU日期截面："
        f"交易日={len(train_date_keys)}，"
        f"平均股票数={cross_section_sizes.mean():.1f}，"
        f"最少={cross_section_sizes.min()}，最多={cross_section_sizes.max()}"
    )

    gru_model = StockGRU(feat_dim, 64, 2).to(DEVICE)
    optimizer_gru = optim.Adam(gru_model.parameters(), lr=0.001)
    criterion_gru = nn.MSELoss()

    best_rank_ic = -np.inf
    best_model_state = None
    best_epoch = 0
    best_rank_ic_metrics = None
    no_improvement_epochs = 0

    for epoch in range(GRU_MAX_EPOCHS):
        gru_model.train()
        total_train_loss = 0.0
        total_train_count = 0

        date_order = np.random.permutation(len(train_date_keys))
        for date_position in date_order:
            date_key = train_date_keys[int(date_position)]
            date_indices = train_date_index_tensors[date_key]

            original_x = x_train.index_select(0, date_indices)
            enhanced_x = x_enhanced.index_select(0, date_indices)
            date_y = y_train.index_select(0, date_indices)

            batch_x = torch.cat([original_x, enhanced_x], dim=0).to(DEVICE)
            batch_y = torch.cat([date_y, date_y], dim=0).to(DEVICE)

            optimizer_gru.zero_grad()
            prediction = gru_model(batch_x)
            loss = criterion_gru(prediction, batch_y)
            loss.backward()
            optimizer_gru.step()

            batch_size = int(batch_x.size(0))
            total_train_loss += float(loss.item()) * batch_size
            total_train_count += batch_size

        train_mse = total_train_loss / max(total_train_count, 1)
        validation_mse, rank_ic_metrics = _evaluate_gru_on_validation(
            gru_model,
            validation_loader,
            validation_dates,
            validation_instruments,
            criterion_gru,
        )
        mean_rank_ic = rank_ic_metrics["mean"]

        improved = (
            np.isfinite(mean_rank_ic)
            and mean_rank_ic > best_rank_ic + RANKIC_MIN_DELTA
        )
        if improved:
            best_rank_ic = float(mean_rank_ic)
            best_epoch = epoch + 1
            best_rank_ic_metrics = dict(rank_ic_metrics)
            no_improvement_epochs = 0
            best_model_state = {
                key: value.detach().cpu().clone()
                for key, value in gru_model.state_dict().items()
            }
        else:
            no_improvement_epochs += 1

        context.logger.info(
            f"[滚动模型{model_id}] GRU epoch={epoch + 1}/{GRU_MAX_EPOCHS}，"
            f"train_MSE={train_mse:.8f}，val_MSE={validation_mse:.8f}，"
            f"val_RankIC={mean_rank_ic:.6f}，"
            f"RankICIR={rank_ic_metrics['ir']:.4f}，"
            f"RankIC胜率={rank_ic_metrics['positive_rate']:.4f}，"
            f"有效截面={rank_ic_metrics['days']}，"
            f"最佳epoch={best_epoch}，未改善轮数={no_improvement_epochs}"
        )

        if no_improvement_epochs >= EARLY_STOP_PATIENCE:
            context.logger.info(
                f"[滚动模型{model_id}] 触发RankIC早停："
                f"连续{EARLY_STOP_PATIENCE}轮未改善"
            )
            break

    if best_model_state is None or best_rank_ic_metrics is None:
        raise RuntimeError(
            f"滚动模型{model_id}验证集未计算出有效RankIC"
        )

    model_bundle = {
        "model_id": model_id,
        "train_start": train_start,
        "training_data_end": training_data_end,
        "effective_start": effective_start,
        "effective_end": effective_end,
        "model_state": best_model_state,
        "train_mean": train_mean.detach().cpu().clone(),
        "train_std": train_std.detach().cpu().clone(),
        "feat_dim": feat_dim,
        "best_epoch": best_epoch,
        "best_validation_rank_ic": best_rank_ic,
        "validation_rank_ic_ir": best_rank_ic_metrics["ir"],
        "validation_rank_ic_positive_rate": best_rank_ic_metrics[
            "positive_rate"
        ],
        "validation_start_date": pd.Timestamp(validation_start_date),
        "train_stock_count": len(train_stocks),
    }

    context.logger.info(
        f"[滚动模型{model_id}] 训练完成："
        f"最佳epoch={best_epoch}，最佳验证RankIC={best_rank_ic:.6f}，"
        f"适用期={effective_start.date()}~{effective_end.date()}"
    )

    del (
        price_df,
        x_flat,
        x_train,
        x_validation,
        x_enhanced,
        y_train,
        y_validation,
        train_dates,
        train_date_groups,
        train_date_keys,
        train_date_index_tensors,
        cross_section_sizes,
        generator,
        discriminator,
        gru_model,
        validation_dates,
        validation_instruments,
    )
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return model_bundle


# =============================================================================
# 6. BigTrader原生日频回调
# =============================================================================

def _find_rolling_model(context, current_date):
    """按当前日期寻找唯一有效的滚动模型包。"""
    current_ts = pd.Timestamp(current_date).normalize()
    for bundle in context.rolling_models:
        if (
            bundle["effective_start"] <= current_ts
            <= bundle["effective_end"]
        ):
            return bundle
    return None


def _activate_rolling_model(context, model_bundle):
    """仅在模型有效期发生切换时加载对应GRU与标准化参数。"""
    model_id = int(model_bundle["model_id"])
    if context.active_model_id == model_id and context.gru_model is not None:
        return

    if context.gru_model is not None:
        del context.gru_model
        context.gru_model = None
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    gru_model = StockGRU(
        int(model_bundle["feat_dim"]),
        64,
        2,
    )
    gru_model.load_state_dict(model_bundle["model_state"])
    gru_model.to(DEVICE)
    gru_model.eval()

    context.gru_model = gru_model
    context.train_mean = model_bundle["train_mean"]
    context.train_std = model_bundle["train_std"]
    context.active_model_id = model_id

    context.logger.info(
        f"切换至滚动模型{model_id}："
        f"训练窗口={model_bundle['train_start'].date()}~"
        f"{model_bundle['training_data_end'].date()}，"
        f"有效期={model_bundle['effective_start'].date()}~"
        f"{model_bundle['effective_end'].date()}，"
        f"最佳epoch={model_bundle['best_epoch']}，"
        f"验证RankIC={model_bundle['best_validation_rank_ic']:.6f}"
    )


def initialize(context: bigtrader.IContext):
    context.set_commission(
        bigtrader.PerOrder(
            buy_cost=0.0003,
            sell_cost=0.0013,
            min_cost=5,
        )
    )

    context.model_ready = False
    context.gru_model = None
    context.train_mean = None
    context.train_std = None
    context.feat_dim = len(FEATURE_COLUMNS)
    context.last_rebalance_date = None
    context.active_model_id = None

    context.index_weights_df = PRELOADED_INDEX_WEIGHTS.copy()
    context.using_official_weights = PRELOADED_USING_OFFICIAL_WEIGHTS
    context.rolling_plan = [dict(item) for item in PRELOADED_ROLLING_PLAN]

    context.logger.info(
        f"已加载滚动计划：模型数={len(context.rolling_plan)}，"
        f"训练窗口={TRAIN_WINDOW_MONTHS}个月，"
        f"模型有效期={MODEL_VALID_MONTHS}个月，"
        f"数据股票并集={len(ROLLING_DATA_UNIVERSE)}，"
        f"回测股票并集={len(BACKTEST_UNIVERSE)}，"
        f"使用官方权重={context.using_official_weights}"
    )

    # 一次性读取所有滚动训练和回测预测需要的原始日线，避免每个模型重复查询DAI。
    context.logger.info(
        f"一次性读取滚动日线缓存："
        f"{ROLLING_DAILY_START}~{BACKTEST_END}，"
        f"股票并集={len(ROLLING_DATA_UNIVERSE)}"
    )
    daily_df = _query_daily_bars(
        ROLLING_DATA_UNIVERSE,
        ROLLING_DAILY_START,
        BACKTEST_END,
    )
    if daily_df.empty:
        raise RuntimeError("滚动训练与回测日线缓存为空")

    context.daily_raw_by_stock = {
        stock: group.sort_values("date").reset_index(drop=True)
        for stock, group in daily_df.groupby("instrument", sort=False)
    }
    del daily_df
    gc.collect()

    # 按自动计划依次训练所有模型；每个模型只能访问其生效日前的数据。
    context.rolling_models = []
    for model_spec in context.rolling_plan:
        model_bundle = _train_single_rolling_model(context, model_spec)
        context.rolling_models.append(model_bundle)

    context.model_ready = len(context.rolling_models) > 0
    if not context.model_ready:
        raise RuntimeError("未成功训练任何滚动模型")

    context.logger.info(
        f"全部滚动模型训练完成：共{len(context.rolling_models)}个。"
        "BigTrader回测期间将按日期自动切换模型。"
    )


def _build_realtime_factor_data(context, stocks, current_date):
    """
    复刻原版：
        get_price(stocks, end_date=current_date,
                  count=SEQ_LEN+60, frequency='daily', fq='pre')

    日频回测在周五调仓，只使用current_date之前已经完整结束的日线。
    因此信号在周五交易前已经可得，不使用周五收盘后信息。
    """
    cutoff = pd.Timestamp(current_date)
    pieces = []

    for stock in stocks:
        group = context.daily_raw_by_stock.get(stock)
        if group is None or group.empty:
            continue

        # 原版count=100；从完整日线中取最近100条。
        section = group[group["date"] < cutoff].tail(DAILY_LOOKBACK_COUNT).copy()
        if section.empty:
            continue
        pieces.append(section)

    if not pieces:
        return pd.DataFrame()

    realtime_df = pd.concat(pieces, ignore_index=True)
    realtime_df = _convert_to_endpoint_pre_adjusted(realtime_df)
    realtime_df, feat_cols = calculate_all_factors(realtime_df)
    realtime_df.dropna(inplace=True)
    return realtime_df, feat_cols


def _predict_scores(context, stocks, current_date):
    realtime_result = _build_realtime_factor_data(
        context,
        stocks,
        current_date,
    )
    if isinstance(realtime_result, pd.DataFrame):
        return {}

    realtime_df, feat_cols = realtime_result
    if realtime_df.empty:
        return {}

    # 批量推理只减少Python循环和模型调用次数，不改变每只股票的输入、
    # 标准化方式、模型结构或预测结果含义。预测值对应未来5日累计收益，实时推理仍不经过Generator。
    valid_stocks = []
    input_list = []
    for stock, group in realtime_df.groupby("instrument", sort=False):
        if len(group) < SEQ_LEN:
            continue
        valid_stocks.append(str(stock))
        input_list.append(group[feat_cols].values[-SEQ_LEN:])

    if not input_list:
        return {}

    input_tensor = torch.tensor(
        np.asarray(input_list, dtype=np.float32),
        dtype=torch.float32,
    )
    input_tensor = (input_tensor - context.train_mean) / context.train_std

    with torch.no_grad():
        scores = (
            context.gru_model(input_tensor.to(DEVICE))
            .detach()
            .cpu()
            .numpy()
            .reshape(-1)
        )

    return {stock: float(score) for stock, score in zip(valid_stocks, scores)}


def _build_target_weights(index_weights, predict_scores):
    # 预测分数从高到低排序。
    sorted_stocks = sorted(
        predict_scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    total_valid = len(sorted_stocks)
    group_size = total_valid // 5

    bias_map = {}
    for idx, (stock, _) in enumerate(sorted_stocks):
        if idx < group_size:
            bias_map[stock] = BIAS
        elif idx < group_size * 2:
            bias_map[stock] = BIAS * 0.5
        elif idx < group_size * 3:
            bias_map[stock] = 0.0
        elif idx < group_size * 4:
            bias_map[stock] = -BIAS * 0.5
        else:
            bias_map[stock] = -BIAS

    target_weights = {}
    for row in index_weights.itertuples(index=False):
        stock = str(row.instrument)
        index_weight = float(row.weight)

        if stock in bias_map:
            target_weight = index_weight + bias_map[stock]
        else:
            target_weight = index_weight

        target_weights[stock] = max(0.0, target_weight)

    total_weight = sum(target_weights.values())
    if total_weight > 0:
        for stock in target_weights:
            target_weights[stock] /= total_weight

    return target_weights


def _execute_rebalance(context, target_weights):
    # A. 清仓不在本期组合内或目标权重为0的股票。
    for stock, position in context.get_account_positions().items():
        current_qty = getattr(position, "current_qty", 0)
        if current_qty <= 0:
            continue
        if stock not in target_weights or target_weights[stock] == 0:
            context.order_target_percent(stock, 0)

    # B. 使用BigTrader原生目标权重订单进行多退少补。
    for stock, target_weight in target_weights.items():
        if target_weight > 0:
            context.order_target_percent(stock, target_weight)


def handle_data(context: bigtrader.IContext, data: bigtrader.IBarData):
    if not context.model_ready:
        return

    now = data.current_dt

    # 日频引擎每个交易日只触发一次；仍然仅在周五调仓。
    if now.weekday() != 4:
        return

    current_date = now.strftime("%Y-%m-%d")
    if context.last_rebalance_date == current_date:
        return

    model_bundle = _find_rolling_model(context, current_date)
    if model_bundle is None:
        context.logger.warning(f"{current_date}没有对应的滚动模型，跳过调仓")
        context.last_rebalance_date = current_date
        return
    _activate_rolling_model(context, model_bundle)

    index_weights = _weights_asof(context.index_weights_df, current_date)
    if index_weights.empty:
        context.logger.warning(f"{current_date}未取得沪深300权重，跳过调仓")
        context.last_rebalance_date = current_date
        return

    stocks = index_weights["instrument"].astype(str).tolist()
    predict_scores = _predict_scores(context, stocks, current_date)
    if not predict_scores:
        context.logger.warning(f"{current_date}无有效预测分数，跳过调仓")
        context.last_rebalance_date = current_date
        return

    target_weights = _build_target_weights(index_weights, predict_scores)
    if not target_weights:
        context.logger.warning(f"{current_date}目标权重为空，跳过调仓")
        context.last_rebalance_date = current_date
        return

    _execute_rebalance(context, target_weights)
    context.last_rebalance_date = current_date

    context.logger.info(
        f"{current_date}日频调仓完成："
        f"指数成分={len(index_weights)}，"
        f"有效预测={len(predict_scores)}，"
        f"使用滚动模型={context.active_model_id}，"
        f"使用官方权重={context.using_official_weights}"
    )


# =============================================================================
# 7. 滚动计划、指数权重和股票池静态准备
# =============================================================================

def _prepare_rolling_static_data():
    """启动BigTrader前自动生成模型计划，并准备权重与股票并集。"""
    rolling_plan = _build_rolling_model_plan()
    if not rolling_plan:
        raise RuntimeError("自动滚动模型计划为空")

    earliest_train_start = min(
        item["train_start"] for item in rolling_plan
    )
    weight_start = (
        earliest_train_start - pd.Timedelta(days=180)
    ).strftime("%Y-%m-%d")

    try:
        weights = _query_official_index_weights(
            weight_start,
            BACKTEST_END,
        )
        if weights.empty:
            raise RuntimeError("官方权重查询结果为空")
        using_official = True
    except Exception as exc:
        print(
            "官方指数权重读取失败，按兼容逻辑退化为沪深300成分股等权。"
            f"原因：{exc}"
        )
        weights = _query_equal_weight_hs300(
            weight_start,
            BACKTEST_END,
        )
        if weights.empty:
            raise RuntimeError(
                "官方权重与等权回退均未取得数据"
            ) from exc
        using_official = False

    # DAI日线缓存需要覆盖所有滚动训练窗口和回测期间可能出现的成分股。
    rolling_rows = weights.loc[
        (weights["date"] >= earliest_train_start - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    rolling_data_universe = sorted(
        rolling_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not rolling_data_universe:
        raise RuntimeError("滚动训练与回测股票并集为空")

    # BigTrader只需加载回测期间可能交易的成分股，降低引擎数据准备开销。
    backtest_rows = weights.loc[
        (weights["date"] >= pd.Timestamp(BACKTEST_START) - pd.Timedelta(days=180))
        & (weights["date"] <= pd.Timestamp(BACKTEST_END))
    ]
    backtest_universe = sorted(
        backtest_rows["instrument"].dropna().astype(str).unique().tolist()
    )
    if not backtest_universe:
        raise RuntimeError("回测期沪深300历史成分股并集为空")

    rolling_daily_start = (
        earliest_train_start - pd.Timedelta(days=TRAIN_FETCH_BUFFER_DAYS)
    ).strftime("%Y-%m-%d")

    print(
        f"自动生成滚动模型计划：{len(rolling_plan)}个模型，"
        f"训练窗口={TRAIN_WINDOW_MONTHS}个月，"
        f"模型有效期={MODEL_VALID_MONTHS}个月"
    )
    for item in rolling_plan:
        print(
            f"模型{item['model_id']}："
            f"训练={item['train_start'].date()}~"
            f"{item['training_data_end'].date()}，"
            f"有效={item['effective_start'].date()}~"
            f"{item['effective_end'].date()}"
        )

    return (
        rolling_plan,
        weights,
        using_official,
        backtest_universe,
        rolling_data_universe,
        rolling_daily_start,
    )


(
    PRELOADED_ROLLING_PLAN,
    PRELOADED_INDEX_WEIGHTS,
    PRELOADED_USING_OFFICIAL_WEIGHTS,
    BACKTEST_UNIVERSE,
    ROLLING_DATA_UNIVERSE,
    ROLLING_DAILY_START,
) = _prepare_rolling_static_data()


# =============================================================================
# 8. BigQuant原生BigTrader日频回测
# =============================================================================

performance = bigtrader.run(
    market=bigtrader.Market.CN_STOCK,
    frequency=bigtrader.Frequency.DAILY,
    instruments=BACKTEST_UNIVERSE,
    start_date=BACKTEST_START,
    end_date=BACKTEST_END,
    capital_base=CAPITAL_BASE,
    initialize=initialize,
    handle_data=handle_data,
    benchmark=INDEX_CODE,
    volume_limit=1,
    order_price_field_buy="close",
    order_price_field_sell="close",
    backtest_only=True,
)

```

滚动训练版在回测期内取得了 19.37% 的累计收益和 8.75% 的年化收益，但同期沪深300基准收益达到 28.36%，策略累计跑输基准约 9个百分点。

策略的 Beta 为 0.71，说明整体市场风险暴露低于沪深300，但 Alpha 为 -0.01、信息比率为 -0.52，表明模型没有形成稳定的超额收益。胜率为 48.34%，虽然盈亏比达到 1.51，但选股正确率仍不足以持续战胜基准。

最大回撤为 13.64%，夏普比率仅为 0.45，风险调整后收益表现一般。从收益曲线看，策略在部分下跌阶段具有一定防御作用，但在2025年下半年至2026年的上涨阶段明显落后于基准。

总体来看，滚动更新提高了模型对市场变化的适应能力，但当前模型的股票排序能力仍不稳定，尚未形成持续有效的指数增强效果。
