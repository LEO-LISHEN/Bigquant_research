```python
# -*- coding: utf-8 -*-
"""SVM 候选特征评价：可直接粘贴到 BigQuant Notebook 的一个执行单元。

填写下方配置后直接运行。该单元自行构造 feature_panel 与 label_data，
不会依赖 Notebook 中预先存在的同名变量。
"""

from pathlib import Path
import sys
import time

import numpy as np
import pandas as pd
from IPython.display import display


# 只在这里填写本次研究参数；所有日期均为闭区间。
# information_anchor_date 是研究当时的信息截止日：仅保留在该日以前
# 已完整走完预测标签窗口的样本，避免把尚未实现的未来收益作为标签。
SVM_FEATURE_SCREENING_CONFIG = {
    # 日期必须是带英文双引号的字符串，格式 "YYYY-MM-DD"。
    # 例如："2023-01-03"；不要写成 2023-01-03。
    # 允许填写自然日，代码会自动取区间内的交易日，并将截止日
    # 向前对齐到最近交易日。
    "evaluation_start_date": "2023-01-03",
    "evaluation_end_date": "2023-12-29",
    "information_anchor_date": "2024-02-01",
    # 仅用于查询交易日历的最早日期；通过 BigQuant 日期分区过滤避免扫描整表。
    # 不改变因子数据的实际预热窗口。
    "trading_calendar_start_date": "2017-01-01",
    "signal_interval_trading_days": 20,
    "prediction_label_window_trading_days": 20,
    "min_obs": 30,
    # 任选其一：
    # {"type": "all_a"}
    # {"type": "custom", "instruments": ["000001.SZ", "600000.SH"]}
    # {"type": "index", "index_codes": ["000300.SH"]}
    "universe": {"type": "all_a"},
    # 自动使用 Factor Repository/base_factors 中全部已登记的 candidate_instances。
    # 每个实例严格使用登记参数，不缩短或改写任何时间窗口。
    "plot_top_n": 6,
}

_required_config_keys = [
    "evaluation_start_date",
    "evaluation_end_date",
    "information_anchor_date",
]
_missing_config = [
    key for key in _required_config_keys
    if not SVM_FEATURE_SCREENING_CONFIG.get(key)
]
if _missing_config:
    raise ValueError(
        "请先填写 SVM_FEATURE_SCREENING_CONFIG 中的参数："
        f"{_missing_config}。"
    )
_project_hint = Path(
    globals().get("PROJECT_ROOT", "/home/aiuser/work")
).expanduser()
_project_candidates = [_project_hint, *_project_hint.parents, Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next(
    (
        candidate for candidate in _project_candidates
        if (candidate / "factor_lib").is_dir()
    ),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "未找到包含 factor_lib 的项目根目录。"
        "BigQuant 中通常应为 /home/aiuser/work；"
        "也可在运行前设置 PROJECT_ROOT。"
    )
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from factor_lib.common.data_adapters.bigquant_adapters.daily import (  # noqa: E402
    load_daily_raw_data,
)
from factor_lib.common.data_adapters.bigquant_adapters.loader import (  # noqa: E402
    get_factor_data_requirements,
    load_factor_raw_data,
)
from factor_lib.function.bigquant_function.factor_evaluation.calculate_factor_basic_metrics import (  # noqa: E402
    _calculate_metrics_from_panels,
    _plot_ic_rankic_series,
    _resolve_universe_panel,
)
from factor_lib.function.bigquant_function.factor_evaluation.calculate_factor_correlation import (  # noqa: E402
    _calculate_correlation_from_panel,
)
from factor_lib.factor_hub.get_factor import get_factor  # noqa: E402
from factor_lib.factor_hub.discover_factors import discover_factors  # noqa: E402


def _load_trading_calendar_with_date_filter(start_date, end_date):
    """以日期分区读取交易日历，避免对 cn_stock_bar1d 做无起点扫描。"""
    try:
        import dai
    except ImportError as exc:
        raise ImportError("未能导入 dai；请在 BigQuant 环境运行。") from exc

    start_date = pd.Timestamp(start_date).normalize()
    end_date = pd.Timestamp(end_date).normalize()
    if start_date > end_date:
        raise ValueError("trading_calendar_start_date 不能晚于信息截止日。")
    started_at = time.perf_counter()
    start_text = start_date.strftime("%Y-%m-%d")
    end_text = end_date.strftime("%Y-%m-%d")
    print(
        "[SVM 特征评价] 1/6 交易日历查询开始："
        f"{start_text} 至 {end_text}（已启用日期分区过滤）...",
        flush=True,
    )
    sql = f"""
    SELECT DISTINCT date
    FROM cn_stock_bar1d
    WHERE date >= '{start_text}'
      AND date <= '{end_text}'
    ORDER BY date
    """
    calendar = dai.query(
        sql,
        filters={"date": [start_text, end_text]},
    ).df()
    if calendar.empty or "date" not in calendar.columns:
        raise ValueError("未读取到有效的 A 股交易日历。")
    dates = pd.DatetimeIndex(pd.to_datetime(calendar["date"], errors="coerce"))
    if dates.isna().any():
        raise ValueError("交易日历中存在无效日期。")
    dates = dates.normalize().unique().sort_values()
    if dates.empty:
        raise ValueError("A 股交易日历为空。")
    print(
        "[SVM 特征评价] 1/6 交易日历查询完成："
        f"{len(dates)} 个交易日，耗时 {time.perf_counter() - started_at:.1f}s。",
        flush=True,
    )
    return dates


_started_at = time.perf_counter()
_config = SVM_FEATURE_SCREENING_CONFIG
_evaluation_start = pd.Timestamp(_config["evaluation_start_date"]).normalize()
_evaluation_end = pd.Timestamp(_config["evaluation_end_date"]).normalize()
_anchor_date = pd.Timestamp(_config["information_anchor_date"]).normalize()
_calendar_start = pd.Timestamp(
    _config.get("trading_calendar_start_date", "2000-01-01")
).normalize()
_signal_interval = int(_config["signal_interval_trading_days"])
_label_window = int(_config["prediction_label_window_trading_days"])
_min_obs = int(_config["min_obs"])
_universe_config = dict(_config["universe"])
_plot_top_n = int(_config["plot_top_n"])

if _evaluation_start > _evaluation_end:
    raise ValueError("evaluation_start_date 不能晚于 evaluation_end_date。")
if _calendar_start > _evaluation_start:
    raise ValueError(
        "trading_calendar_start_date 不能晚于 evaluation_start_date；"
        "否则无法保证因子预热所需的历史交易日历。"
    )
if _anchor_date < _evaluation_end:
    raise ValueError(
        "information_anchor_date 不能早于 evaluation_end_date；"
        "它是本次研究可以使用信息的最终截止日。"
    )
if _signal_interval <= 0 or _label_window <= 0 or _min_obs <= 1:
    raise ValueError("信号间隔、预测标签窗口必须为正，min_obs 至少为 2。")
if _universe_config.get("type") not in {"all_a", "custom", "index"}:
    raise ValueError("universe.type 仅支持 all_a、custom 或 index。")

_base_factor_metadata = discover_factors()
_reserved_factor_params = {
    "target_dates", "as_of_date", "show_progress", "progress_every"
}
_factor_specs = []
# 部分较早的基础因子尚未补充 candidate_instances 元数据。它们仍是
# 可用的基础因子，因此按 FACTOR.parameters 中登记的默认参数作为一个
# 单独实例纳入；不会修改、缩短或猜测其参数。
_default_instance_fallback_factors = []
for _base_factor_name, _metadata in sorted(_base_factor_metadata.items()):
    if _metadata.get("factor_type") != "base":
        continue
    _candidate_instances = _metadata.get("candidate_instances")
    _instance_source = "registered_candidate_instances"
    if not isinstance(_candidate_instances, dict) or not _candidate_instances:
        _candidate_instances = {"default": {}}
        _instance_source = "metadata_default_fallback"
        _default_instance_fallback_factors.append(_base_factor_name)
    _metadata_parameters = _metadata.get("parameters", {})
    _default_factor_params = {
        _parameter_name: _parameter_spec["default"]
        for _parameter_name, _parameter_spec in _metadata_parameters.items()
        if isinstance(_parameter_spec, dict)
        and "default" in _parameter_spec
        and _parameter_name not in _reserved_factor_params
    }
    for _instance_name, _instance_params in sorted(_candidate_instances.items()):
        if not isinstance(_instance_name, str) or not _instance_name.strip():
            raise ValueError(
                f"因子 {_base_factor_name!r} 存在空的 candidate_instances 名称。"
            )
        if not isinstance(_instance_params, dict):
            raise TypeError(
                f"因子 {_base_factor_name!r} 的候选实例 {_instance_name!r} "
                "参数必须是字典。"
            )
        _conflicting_instance_params = sorted(
            set(_instance_params) & _reserved_factor_params
        )
        if _conflicting_instance_params:
            raise ValueError(
                f"因子 {_base_factor_name!r} 的候选实例 {_instance_name!r} "
                f"不能登记运行控制参数：{_conflicting_instance_params}。"
            )
        _factor_params = dict(_default_factor_params)
        _factor_params.update(_instance_params)
        _feature_name = (
            _base_factor_name
            if _instance_name == "default"
            else f"{_base_factor_name}_{_instance_name}"
        )
        _factor_specs.append(
            {
                "feature_name": _feature_name,
                "factor_name": _base_factor_name,
                "candidate_instance": _instance_name,
                "instance_source": _instance_source,
                "params": _factor_params,
            }
        )
_feature_names = [item["feature_name"] for item in _factor_specs]
if not _feature_names:
    raise ValueError(
        "没有发现任何 factor_type='base' 的基础因子。"
    )
if len(set(_feature_names)) != len(_feature_names):
    raise ValueError("candidate_instances 生成了重复的特征实例名称。")
_factor_plan = pd.DataFrame(
    [
        {
            "feature_name": item["feature_name"],
            "factor_name": item["factor_name"],
            "candidate_instance": item["candidate_instance"],
            "instance_source": item["instance_source"],
            "params": item["params"],
        }
        for item in _factor_specs
    ]
)
print(
    f"[SVM 特征评价] 已自动发现 {len(_factor_specs)} 个基础因子候选实例；"
    "以下为本次实际参数：",
    flush=True,
)
display(_factor_plan)
if _default_instance_fallback_factors:
    print(
        "[SVM 特征评价] 注意：以下基础因子尚未登记 candidate_instances，"
        "已按 FACTOR.parameters 的默认参数各纳入一个实例（未改写参数）："
        f"{_default_instance_fallback_factors}",
        flush=True,
    )

_calendar_end = max(_evaluation_end, _anchor_date)
trading_calendar = _load_trading_calendar_with_date_filter(
    _calendar_start,
    _calendar_end,
)
_calendar_positions = {date: position for position, date in enumerate(trading_calendar)}
_effective_anchor_candidates = trading_calendar[trading_calendar <= _anchor_date]
if _effective_anchor_candidates.empty:
    raise ValueError("information_anchor_date 早于可用交易日历。")
_effective_anchor_date = _effective_anchor_candidates.max()

_candidate_signal_dates = trading_calendar[
    (_evaluation_start <= trading_calendar)
    & (trading_calendar <= _evaluation_end)
][::_signal_interval]
_anchor_position = _calendar_positions[_effective_anchor_date]
signal_dates = pd.DatetimeIndex([
    date for date in _candidate_signal_dates
    if _calendar_positions[date] + _label_window + 1 <= _anchor_position
])
if signal_dates.empty:
    raise ValueError(
        "没有标签窗口已完成的信号日；请延后 information_anchor_date、"
        "缩短 prediction_label_window_trading_days 或前移评价区间。"
    )

_schedule = pd.DataFrame({"date": signal_dates})
_schedule["entry_date"] = [
    trading_calendar[_calendar_positions[date] + 1]
    for date in signal_dates
]
_schedule["return_end_date"] = [
    trading_calendar[_calendar_positions[date] + _label_window + 1]
    for date in signal_dates
]
_price_dates = pd.DatetimeIndex(
    pd.concat([
        pd.Series(_schedule["date"]),
        pd.Series(_schedule["entry_date"]),
        pd.Series(_schedule["return_end_date"]),
    ])
    .drop_duplicates()
    .sort_values()
)

print("[SVM 特征评价] 2/6 读取历史动态股票池...", flush=True)
_universe_config, universe_panel, _load_instruments = _resolve_universe_panel(
    _universe_config,
    signal_dates,
)

print("[SVM 特征评价] 3/6 读取开盘价并构造连续超额收益标签...", flush=True)
_price_panel = load_daily_raw_data(
    ["open"],
    dates=_price_dates,
    instruments=_load_instruments,
    show_progress=True,
)
_price_panel["date"] = pd.to_datetime(_price_panel["date"]).dt.normalize()
if universe_panel is None:
    # all_a 的资格以信号日当日可查询到的股票日频记录确定；不会使用未来日期补池。
    universe_panel = _price_panel.loc[
        _price_panel["date"].isin(signal_dates),
        ["date", "instrument"],
    ].drop_duplicates()
else:
    universe_panel = universe_panel[["date", "instrument"]].copy()
    universe_panel["date"] = pd.to_datetime(universe_panel["date"]).dt.normalize()

label_data = universe_panel.merge(_schedule, on="date", how="left", validate="many_to_one")
_entry_prices = _price_panel.rename(
    columns={"date": "entry_date", "open": "entry_open"}
)[["entry_date", "instrument", "entry_open"]]
_exit_prices = _price_panel.rename(
    columns={"date": "return_end_date", "open": "exit_open"}
)[["return_end_date", "instrument", "exit_open"]]
label_data = label_data.merge(
    _entry_prices,
    on=["entry_date", "instrument"],
    how="left",
    validate="one_to_one",
).merge(
    _exit_prices,
    on=["return_end_date", "instrument"],
    how="left",
    validate="one_to_one",
)
label_data["raw_forward_return"] = (
    label_data["exit_open"] / label_data["entry_open"] - 1.0
)
label_data.loc[
    ~np.isfinite(label_data["raw_forward_return"]),
    "raw_forward_return",
] = np.nan
label_data["forward_return"] = label_data["raw_forward_return"] - label_data.groupby(
    "date"
)["raw_forward_return"].transform("mean")
label_data = label_data[
    ["date", "instrument", "entry_date", "return_end_date", "forward_return"]
].sort_values(["date", "instrument"]).reset_index(drop=True)

print("[SVM 特征评价] 4/6 使用现有 loader/get_factor 计算候选因子...", flush=True)
feature_panel = universe_panel.sort_values(["date", "instrument"]).reset_index(drop=True)
for _position, _spec in enumerate(_factor_specs, start=1):
    _feature_name = _spec["feature_name"].strip()
    _factor_name = _spec["factor_name"].strip()
    _factor_params = dict(_spec.get("params", {}))
    _reserved_factor_params = {
        "target_dates", "as_of_date", "show_progress", "progress_every"
    }
    _conflicting_params = sorted(
        set(_factor_params) & _reserved_factor_params
    )
    if _conflicting_params:
        raise ValueError(
            f"{_feature_name} 的 params 不能包含由本 Notebook 管理的参数："
            f"{_conflicting_params}。"
        )
    print(
        f"[SVM 特征评价] 因子 {_position}/{len(_factor_specs)}："
        f"{_feature_name} <- {_factor_name}",
        flush=True,
    )
    _requirements = get_factor_data_requirements(_factor_name, _factor_params)
    _data_window = _requirements["data_window"]
    _lookback = max(
        int(_data_window.get("lookback_trading_days", 0)),
        int(_data_window.get("minimum_history_observations", 0)),
    )
    _first_position = _calendar_positions[signal_dates.min()]
    _factor_dates = trading_calendar[
        max(0, _first_position - _lookback): _calendar_positions[signal_dates.max()] + 1
    ]
    _raw_data = load_factor_raw_data(
        _factor_name,
        dates=_factor_dates,
        factor_params=_factor_params,
        instruments=_load_instruments,
        show_progress=True,
    )
    _factor_output = get_factor(
        _factor_name,
        _raw_data,
        target_dates=signal_dates,
        as_of_date=signal_dates.max(),
        show_progress=True,
        **_factor_params,
    )
    if not isinstance(_factor_output, pd.DataFrame):
        raise TypeError(f"因子 {_factor_name!r} 的输出必须是 pandas.DataFrame。")
    _value_columns = [
        column for column in _factor_output.columns
        if column not in {"date", "instrument"}
    ]
    if len(_value_columns) != 1:
        raise ValueError(
            f"因子 {_factor_name!r} 应恰好输出一个因子值列，"
            f"实际为：{_value_columns}。"
        )
    _factor_output = _factor_output[["date", "instrument", _value_columns[0]]].copy()
    _factor_output["date"] = pd.to_datetime(_factor_output["date"]).dt.normalize()
    _factor_output = _factor_output.loc[_factor_output["date"].isin(signal_dates)]
    _factor_output = _factor_output.rename(columns={_value_columns[0]: _feature_name})
    feature_panel = feature_panel.merge(
        _factor_output,
        on=["date", "instrument"],
        how="left",
        validate="one_to_one",
    )

print("[SVM 特征评价] 5/6 调用已有基础指标与相关性函数...", flush=True)
_evaluation_panel = feature_panel.merge(
    label_data,
    on=["date", "instrument"],
    how="left",
    validate="one_to_one",
)
_metric_rows = []
_metric_details = {}
for _position, _feature_name in enumerate(_feature_names, start=1):
    print(
        f"[SVM 特征评价] 基础指标 {_position}/{len(_feature_names)}：{_feature_name}",
        flush=True,
    )
    _summary, _timeseries = _calculate_metrics_from_panels(
        factor_data=_evaluation_panel,
        label_data=label_data,
        factor_column=_feature_name,
        min_obs=_min_obs,
        show_progress=True,
        progress_every=20,
        started_at=_started_at,
        stage_number=_position,
        stage_total=len(_feature_names),
    )
    _metric_details[_feature_name] = {
        "summary": _summary,
        "timeseries": _timeseries,
    }
    _row = _summary.iloc[0].to_dict()
    _row["feature_name"] = _feature_name
    _row["coverage"] = float(_evaluation_panel[_feature_name].notna().mean())
    _metric_rows.append(_row)

metrics_summary = pd.DataFrame(_metric_rows).loc[:, [
    "feature_name", "coverage", "ic_mean", "icir", "rank_ic_mean",
    "rank_icir", "factor_return_mean", "mean_t_value",
    "cross_section_count", "valid_ic_count",
]].sort_values("rank_icir", key=lambda values: values.abs(), ascending=False)

if len(_feature_names) >= 2:
    correlation_matrix, correlation_overlap_days = _calculate_correlation_from_panel(
        factor_data=feature_panel,
        factor_names=_feature_names,
        method="spearman",
        min_obs=_min_obs,
        show_progress=True,
        progress_every=20,
        started_at=_started_at,
        stage_number=1,
        stage_total=1,
    )
else:
    correlation_matrix = pd.DataFrame([[1.0]], index=_feature_names, columns=_feature_names)
    correlation_overlap_days = pd.DataFrame([[len(signal_dates)]], index=_feature_names, columns=_feature_names)

print("[SVM 特征评价] 6/6 输出评价表与 RankIC 时序图...", flush=True)
display(metrics_summary)
display(correlation_matrix)
display(correlation_overlap_days)
for _feature_name in metrics_summary["feature_name"].head(_plot_top_n):
    _plot_ic_rankic_series(
        _metric_details[_feature_name]["timeseries"],
        title=f"{_feature_name}：IC 与 RankIC 时序",
        show=True,
    )

screening_result = {
    "config": _config,
    "project_root": str(PROJECT_ROOT),
    "signal_dates": signal_dates,
    "effective_information_anchor_date": _effective_anchor_date,
    "factor_plan": _factor_plan,
    "default_instance_fallback_factors": _default_instance_fallback_factors,
    "universe_panel": universe_panel,
    "feature_panel": feature_panel,
    "label_data": label_data,
    "metrics_summary": metrics_summary,
    "metric_details": _metric_details,
    "correlation_matrix": correlation_matrix,
    "correlation_overlap_days": correlation_overlap_days,
}
print(
    f"[SVM 特征评价] 完成：{len(signal_dates)} 个信号日、"
    f"{len(feature_panel):,} 条股票-日期记录、{len(_feature_names)} 个候选特征，"
    f"有效信息截止日 {_effective_anchor_date:%Y-%m-%d}，"
    f"耗时 {time.perf_counter() - _started_at:.1f}s。",
    flush=True,
)
```

```python
# -*- coding: utf-8 -*-
"""将当前 Notebook 的 SVM 特征评价结果完整导出到 factor_lib/temp_outputs。"""

from datetime import datetime
from pathlib import Path
import json

import pandas as pd
from IPython.display import display


if "screening_result" not in globals():
    raise NameError(
        "未发现 screening_result。请先完整运行 SVM 特征评价单元，再运行本导出单元。"
    )
if not isinstance(screening_result, dict):
    raise TypeError("screening_result 必须是字典。请重新运行 SVM 特征评价单元。")

_required_result_keys = {
    "config",
    "factor_plan",
    "metrics_summary",
    "metric_details",
    "correlation_matrix",
    "correlation_overlap_days",
    "feature_panel",
    "label_data",
    "universe_panel",
}
_missing_result_keys = sorted(_required_result_keys - set(screening_result))
if _missing_result_keys:
    raise KeyError(
        "screening_result 缺少必要结果："
        f"{_missing_result_keys}。请使用最新版评价单元重新运行。"
    )

_project_root = Path(
    screening_result.get("project_root", globals().get("PROJECT_ROOT", "/home/aiuser/work"))
).expanduser().resolve()
_output_root = (_project_root / "factor_lib" / "temp_outputs").resolve()
if _output_root.parent.name != "factor_lib":
    raise ValueError("导出目录必须位于 PROJECT_ROOT/factor_lib/temp_outputs。")

_run_id = datetime.now().strftime("svm_feature_screening_%Y%m%d_%H%M%S")
OUTPUT_DIR = _output_root / _run_id
OUTPUT_DIR.mkdir(parents=True, exist_ok=False)
_saved_files = []


def _save_frame(frame, file_stem, *, include_index=False):
    """优先写 Parquet；环境没有引擎时回退为 UTF-8-SIG 压缩 CSV。"""
    if not isinstance(frame, pd.DataFrame):
        raise TypeError(f"{file_stem} 必须是 pandas.DataFrame。")
    parquet_path = OUTPUT_DIR / f"{file_stem}.parquet"
    try:
        frame.to_parquet(parquet_path, index=include_index)
        saved_path = parquet_path
    except ImportError:
        csv_path = OUTPUT_DIR / f"{file_stem}.csv.gz"
        frame.to_csv(
            csv_path,
            index=include_index,
            encoding="utf-8-sig",
            compression="gzip",
        )
        saved_path = csv_path
    _saved_files.append(saved_path.name)
    return saved_path


def _json_safe(value):
    return json.dumps(value, ensure_ascii=False, default=str, sort_keys=True)


print(f"[SVM 结果导出] 开始写入：{OUTPUT_DIR}", flush=True)

_factor_plan = screening_result["factor_plan"].copy()
if "params" in _factor_plan.columns:
    _factor_plan["params"] = _factor_plan["params"].map(_json_safe)
_save_frame(_factor_plan, "factor_plan")
_save_frame(screening_result["metrics_summary"], "factor_metrics_summary")
_save_frame(
    screening_result["correlation_matrix"],
    "factor_correlation_spearman",
    include_index=True,
)
_save_frame(
    screening_result["correlation_overlap_days"],
    "factor_correlation_overlap_days",
    include_index=True,
)

_metric_timeseries_frames = []
for _feature_name, _detail in screening_result["metric_details"].items():
    _timeseries = _detail.get("timeseries")
    if not isinstance(_timeseries, pd.DataFrame):
        raise TypeError(f"{_feature_name} 的指标时序不是 pandas.DataFrame。")
    _timeseries = _timeseries.copy()
    _timeseries.insert(0, "feature_name", _feature_name)
    _metric_timeseries_frames.append(_timeseries)
_metric_timeseries = pd.concat(
    _metric_timeseries_frames,
    axis=0,
    ignore_index=True,
    sort=False,
)
_save_frame(_metric_timeseries, "factor_metric_timeseries")

# 保存后续特征筛选直接需要的完整面板；不会重新取数或改变任何结果。
_save_frame(screening_result["feature_panel"], "feature_panel")
_save_frame(screening_result["label_data"], "label_data")
_save_frame(screening_result["universe_panel"], "universe_panel")

_metadata = {
    "created_at": datetime.now().isoformat(timespec="seconds"),
    "config": screening_result["config"],
    "effective_information_anchor_date": str(
        screening_result.get("effective_information_anchor_date")
    ),
    "signal_date_count": len(screening_result.get("signal_dates", [])),
    "feature_count": len(screening_result["factor_plan"]),
    "feature_panel_row_count": len(screening_result["feature_panel"]),
    "label_row_count": len(screening_result["label_data"]),
    "default_instance_fallback_factors": screening_result.get(
        "default_instance_fallback_factors", []
    ),
    "files": _saved_files,
}
_metadata_path = OUTPUT_DIR / "run_metadata.json"
_metadata_path.write_text(
    json.dumps(_metadata, ensure_ascii=False, indent=2, default=str),
    encoding="utf-8",
)
_saved_files.append(_metadata_path.name)

_saved_files_table = pd.DataFrame({"file_name": _saved_files})
display(_saved_files_table)
print(
    f"[SVM 结果导出] 完成：{len(_saved_files)} 个文件已保存至 {OUTPUT_DIR}",
    flush=True,
)
```
