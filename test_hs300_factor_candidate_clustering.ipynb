```python
# -*- coding: utf-8 -*-
"""沪深300基础因子候选聚类与组内排序。

本脚本用于在机器学习因子正式回测开始前，先用独立的历史区间筛选
候选基础因子实例。它不会把 GAN-GRU、其他机器学习因子或复合因子
纳入候选池。
"""

import sys
from pathlib import Path

sys.dont_write_bytecode = True

from factor_lib.function.bigquant_function.factor_evaluation.factor_candidate_clustering import (
    cluster_and_rank_factor_candidates,
)


# ===== 研究口径 =====
# 该区间严格早于下一次 GAN-GRU 指数增强回测（2023-04-28 起），
# 因而候选筛选不会使用该回测期的收益标签。
SCREEN_START_DATE = "2020-01-01"
SCREEN_END_DATE = "2023-03-31"

# 每 20 个交易日取一个截面，并以未来 20 个交易日收益作为标签。
FREQUENCY = 20
HOLDING_PERIOD_DAYS = 20
MIN_OBS = 100
CLUSTER_COUNT = 8

# 采用历史点时沪深300成分股；不是固定的当前成分股名单。
UNIVERSE = {
    "type": "index",
    "index_codes": ["000300.SH"],
}

# None 表示自动展开该基础因子在 FACTOR['candidate_instances'] 中
# 登记的全部常用实例。这里仅列入当前已经迁移的普通单因子。
CANDIDATE_SPEC = {
    # 估值、成长与质量
    "book_to_price": None,
    "profit_g_q": None,
    "sales_g_q": None,
    "qfa_roe": None,

    # 华泰波动率/换手率因子
    "bias_std_turn_nd": None,
    "hml_r_std_nm": None,
    "id1_std_nm": None,
    "id2_std_nm": None,

    # 动量与收益率类
    "return_nm": None,
    "wgt_return_nm": None,
    "exp_wgt_return_nm": None,
    "return_nd": None,
    "return_std_nd": None,
    "price_ma_distance_nd": None,
    "rsi_nd": None,
    "williams_r_nd": None,
    "macd_hist_relative": None,

    # 资金流向与量价类
    "mfd_buyamt_nd": None,
    "mfd_sellamt_nd": None,
    "amount_relative_ma_nd": None,
    "volume_relative_ma_nd": None,
    "volume_ma_ratio_nd": None,
    "volume_ratio_nd": None,

    # 日内价格行为类
    "open_gap": None,
    "low_relative_return": None,
    "high_relative_return": None,
    "intraday_return": None,
}

# 用户此前建立的结果目录。显式保存一次完整 JSON，便于之后发给 LLM
# 或在不同 Notebook 中复核；文件内包含原始排序、展示排序、相似度矩阵与配置。
OUTPUT_DIR = Path("factor_lib/factor_management_outputs")
RUN_NAME = "hs300_base_factor_screen_20200101_20230331"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


print("[候选因子管理] 开始：沪深300基础因子实例聚类与组内排序。")

cluster_result = cluster_and_rank_factor_candidates(
    candidate_spec=CANDIDATE_SPEC,
    start_date=SCREEN_START_DATE,
    end_date=SCREEN_END_DATE,
    frequency=FREQUENCY,
    holding_period_days=HOLDING_PERIOD_DAYS,
    cluster_count=CLUSTER_COUNT,
    universe=UNIVERSE,
    min_obs=MIN_OBS,
    # 标签期限与截面间隔相同，因此标签不重叠，HAC 滞后阶数设为 0。
    hac_lags=0,
    significance_threshold=2.0,
    plot=True,
    display_results=True,
    save_result=True,
    output_dir=OUTPUT_DIR,
    run_name=RUN_NAME,
    show_progress=True,
    progress_every=1,
)

print("\n===== 候选因子管理完成 =====")
print(f"候选实例数：{len(cluster_result['candidate_ranking'])}")
print(f"聚类组数：{len(cluster_result['group_summary'])}")
print(f"研究区间：{SCREEN_START_DATE} 至 {SCREEN_END_DATE}")
print(f"结果文件：{cluster_result['saved_result_path']}")

# 以下三个对象会保留在 Notebook 内存中，便于你将输出贴给我进一步挑选特征。
group_summary = cluster_result["group_summary"]
group_ranking = cluster_result["group_ranking"]
group_ranking_display = cluster_result["group_ranking_display"]

display(group_summary)
display(group_ranking_display)
```
